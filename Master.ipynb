````python
# Time calculation
from datetime import datetime, timedelta

def working_hours(input_):
    start = []
    end = []
    for i in input_.strip().split('\n'):
        if 'In' in i:
            start.append(i.replace('In','').strip())
        if 'Out' in i:
            end.append(i.replace('Out','').strip())

    time_now = datetime.now()
    total_sec = 0
    for i,j in zip(start, end):
        t1 = datetime.combine(time_now.date(),datetime.strptime(i, '%I:%M:%S %p').time())
        t2 = datetime.combine(time_now.date(),datetime.strptime(j, '%I:%M:%S %p').time())
        total_sec += (t2-t1).total_seconds()
    if len(start) > len(end):
        last_in = datetime.combine(time_now.date(),datetime.strptime(start[-1], '%I:%M:%S %p').time())
        total_sec += (time_now - last_in).total_seconds()
    print(f"Total working hours: {total_sec // 3600} hours {(total_sec % 3600) // 60} minutes")
    remaining = (((8*3600)+(30*60))-total_sec)
    print(f'Remaining: {remaining//3600} hours left {(remaining%3600)//60} minutes left')
    r_hours,r_minutes = remaining // 3600,(remaining % 3600) // 60
    remaining = timedelta(hours=r_hours, minutes=r_minutes)
    now = datetime.now()
    end_time = now + remaining
    print("End time:", end_time.strftime('%I:%M:%S %p'))
    
input_ = '''

10:33:13 AM	In
02:19:53 PM	Out
02:29:01 PM	In
'''
working_hours(input_)
````

````python
# Auto rotating
import cv2
import numpy as np
from PIL import Image

def display_image(image):
    display(Image.fromarray(image))
    
def rotated(image):
    img = cv2.imread(image)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)
    lines = cv2.HoughLines(edges, 1, np.pi/180, 200)
    angles = []
    for rho, theta in lines[:,0]:
        angle = (theta * 180 / np.pi) - 90
        angles.append(angle)
    angle = np.median(angles)
    (h, w) = img.shape[:2]
    center = (w//2, h//2)
    M = cv2.getRotationMatrix2D(center, angle, 1.0)
    return cv2.warpAffine(img, M, (w, h),flags=cv2.INTER_CUBIC,borderMode=cv2.BORDER_REPLICATE)

filename = r"C:\Users\peter.leo\Downloads\Projects\OCR\Blob_storage_ACE_Model\Blob_storage\uae_driving_front_v2\sample_49.png"
raw_image = cv2.imread(filename)
display_image(raw_image)
rotated_image = rotated(filename)
display_image(rotated_image)
````

````python
# regex_doc_expiry
from datetime import datetime
import re

def doc_expiry(raw_text):
    dates = []
    for text in raw_text:
        matches = re.findall(r'\b\d{2}[-/]?\d{2}[-/]?\d{4}\b', text)
        for m in matches:
            clean = re.sub(r'[-/]', '', m)
            try:
                dt = datetime.strptime(clean,"%d%m%Y")
                dates.append(dt)
            except:
                pass

    max_date, today = max(dates), datetime.today()
    return 'Valid' if max_date > today else 'Expired'

doc_expiry(raw_text)
````

````python
# Image to uploadfile for yolo splitter
import cv2, io, pypdfium2
import numpy as np
from importlib.resources import files
from fastapi import UploadFile
from ultralytics import YOLO
from fastapi import UploadFile
from PIL import Image

# model_path = files("services").joinpath("yolo-multipage-OCR-cls.pt")

moedl_path = r"C:\Users\peter.leo\Downloads\Projects\OCR\UAE_31-03-2026\yolo-multipage-OCR-cls.pt"
model = YOLO(str(moedl_path))

async def extract_documents(results, image):
    output = []
    for r in results:
        if r.boxes is None:continue
        boxes = r.boxes.xyxy.cpu().numpy()
        classes = r.boxes.cls.cpu().numpy()

        for box, cls_id in zip(boxes, classes):
            class_name = model.names[int(cls_id)]
            x1, y1, x2, y2 = map(int, box)
            crop = image[y1:y2, x1:x2]
            ok, buffer = cv2.imencode(".jpg", crop)
            if not ok:continue
            output.append({
                "class": class_name,
                "image_bytes": buffer.tobytes()
            })
    if not output:
        ok, buffer = cv2.imencode(".jpg", image)
        if ok:
            output.append({
                "class": 'doc',
                "image_bytes": buffer.tobytes() })
    return output

async def yolo_split(image_path: UploadFile) -> list[bytes]:
    data = []
    CONFIDENCE_THRESHOLD = 0.5
    IOU_THRESHOLD = 0.4
    img_bytes = await image_path.read()
    if image_path.filename.lower().endswith(".pdf"):
        pdf = pypdfium2.PdfDocument(img_bytes)
        for page in pdf:
            pil_image = page.render(scale=2).to_pil()
            image = cv2.cvtColor(np.array(pil_image), cv2.COLOR_RGB2BGR)
            
            # results = model(image, iou=0.9,verbose=False)
            results = model(image, conf=CONFIDENCE_THRESHOLD, iou=IOU_THRESHOLD, verbose=False)
            data.extend(await extract_documents(results, image))
        pdf.close()
    else:
        nparr = np.frombuffer(img_bytes, np.uint8)
        image = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
        results = model(image, conf=CONFIDENCE_THRESHOLD, iou=IOU_THRESHOLD, verbose=False)
        # results = model(image, iou=0.9, verbose=False)
        data.extend(await extract_documents(results, image))
    return data



def create_upload_file(path):
    with open(path, "rb") as f:
        return UploadFile(
            filename=path.split("\\")[-1],
            file=io.BytesIO(f.read())
        )
        
upload_file = create_upload_file(filename)
files = await yolo_split(upload_file)
Image.open(filename)

ocr_values = []
for file in files:
    poller = azure_client.begin_analyze_document(
        "prebuilt-read",
        body=file['image_bytes'],
    )
    analyze_result = poller.result()
    ocr_text = analyze_result.content
    ocr_values.append(ocr_text)
print(ocr_values[0])
````

````python
# SLM + Preprocessing using co-ords line matching concept

from PIL import Image
def is_arabic(text):
    return re.search(r'[\u0600-\u06FF\u0750-\u077F\u08A0-\u08FF\uFB50-\uFDFF\uFE70-\uFEFF]',text)
def only_english(ocr_list):
    cleaned_text = []
    for d in ocr_list:
        text = ''.join([i for i in d if not is_arabic(i)]).strip()
        if text:cleaned_text.append(text)
    return cleaned_text    

def combine_text_lines(data, threshold=10):

    lines = []

    for item in data:
        poly = item['polygon']
        x = poly[0::2]
        y = poly[1::2]

        center_x = sum(x)/4
        center_y = sum(y)/4

        placed = False

        for line in lines:
            if abs(line['y'] - center_y) < threshold:
                line['words'].append((center_x,item['content']))
                placed = True
                break

        if not placed:
            lines.append({
                "y":center_y,
                "words":[(center_x,item['content'])]
            })

    result = []

    for line in sorted(lines,key=lambda l:l['y']):
        words = sorted(line['words'],key=lambda w:w[0])
        text = " ".join(w[1] for w in words)
        result.append(text)

    return result

filename = r"C:\Users\peter.leo\Downloads\Projects\OCR\Blob_storage_ACE_Model\Blob_storage\uae_driving_front_v2\sample_49.png"
display(Image.open(filename))
data = azure_data(filename).pages[0].lines
combined_lines = combine_text_lines(data)

cleaned_text = only_english(combined_lines)
f = 'license authority id|License No.|Name|Nationality|Date of Birth|Issue Date|Expiry Date|Place of Issue'
fields = f.split('|')


output_format = {i:"" for i in fields}
system = {
 "role": "system",
 "content": f"""
You are a document field extraction system.
Extract the requested fields from OCR text.
Schema:
{output_format}
Strict: Return in JSON Format only
"""
}
prompt = {
 "role": "user",
 "content": f""" OCR TEXT:\nExtract fields :- {cleaned_text}

"""
}
from openai import OpenAI
#Qwen3.5
client = OpenAI(base_url="http://172.20.132.97:11434/v1",api_key="none") # Phi4
# client = OpenAI(base_url="http://localhost:11434/v1",api_key="ollama") # Local Phi3
# client = OpenAI(base_url='http://172.20.132.97:11434/v1',api_key='na') #GLM
 
resp = client.chat.completions.create(
    model="qwen",
    # model='phi3:mini',
    # model='glm-ocr:latest',
    temperature=0,
    response_format={"type": "json_schema","json_schema": {"name": "document_extraction","strict": True}},
    messages=[system,prompt])
 
text = resp.choices[0].message.content
print(text)
# json.loads(text)
````

````python
# Final document cv + Qwen

import cv2
import matplotlib.pyplot as plt
import numpy as np
import os
import json
from dotenv import load_dotenv
load_dotenv()
import base64
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
import json

llm =  ChatOpenAI(api_key=os.getenv("QWEN_KEY"),base_url=os.getenv("QWEN_BASE_URL"),model="Qwen3-VL-4B-Instruct",temperature=0,timeout=60,max_retries=1)


load = []
for image_path in paths:
    image = cv2.imread(image_path)

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (5, 5), 0)
    # edge detection
    edged = cv2.Canny(blur, 75, 200)

    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (50, 50))
    edged = cv2.morphologyEx(edged, cv2.MORPH_CLOSE, kernel)
    
    contours, hierarchy = cv2.findContours(
        edged,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    if len(contours) == 0:
        raise ValueError("No contours found")

    contours = sorted(contours, key=cv2.contourArea, reverse=True)

    MIN_AREA = 0.05 * image.shape[0] * image.shape[1]  # 5% of image area

    canvas = image.copy()
    
    polygons = []
    
    polygon_id = 1

    for document_contour in contours :
        area = cv2.contourArea(document_contour)
        if area < MIN_AREA:
            continue
        rect = cv2.minAreaRect(document_contour)
        (cx, cy), (rw, rh), angle = rect
        box = cv2.boxPoints(rect)
        box = np.int32(box)
        
        polygons.append({
        "polygon_id": polygon_id,
        "points": box.tolist()
            })

        cv2.drawContours(canvas, [box], -1, (0, 0, 255), 2)
        
        cv2.putText(
        canvas,
        f"#{polygon_id}",
        tuple(box[0]),
        cv2.FONT_HERSHEY_SIMPLEX,
        1.2,
        (255, 0, 0),
        3
    )

        polygon_id += 1
        
    canvas_rgb = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8, 10))
    plt.imshow(canvas_rgb)
    plt.title("Detected Document Contour for "+image_path)
    plt.axis("off")
    plt.show()
    
    def image_to_base64(image):
        _, buffer = cv2.imencode('.jpg', image)
        return base64.b64encode(buffer).decode("utf-8")
    b64 = image_to_base64(image)

    initial_prompt = f"""
    You are a document classification engine.

    Each polygon represents ONE document in the image.
    Polygon metadata is provided below.

    For EACH polygon_id:
    
    1. Identify the document type.
    2. Identify the document side, Use these below STRICT rules:
        - BACK:
        If ANY of the following are present:
        - Barcode or QR code
        - MRZ-like text
        - Passport number
        - Serial number
        - Issue / Expiry table
        - Signature block

        - FRONT:
        ONLY if:
        - Large portrait photo is present AND
        - Name, DOB, nationality are present AND
        - NONE of the BACK indicators appear

        If both FRONT and BACK indicators appear,always try to classify them.
    3. Determine if two polygons belong to the SAME physical document.

    POLYGONS:
    {json.dumps(polygons, indent=2)}

    DOCUMENT TYPES:
    - DRIVING_LICENSE
    - VEHICLE_LICENSE
    - ID_CARD
    - RESIDENCY_PERMIT
    - UNKNOWN

    Output strict JSON:
    {{
    "total_documents": <int>,
    "documents": [
        {{
        "polygon_id": <int>,
        "document_type": "<TYPE>",
        "document_side": "<FRONT|BACK>",
        "confidence": <float>
        }}
    ]
    }}
    """
    
    message = HumanMessage(
        content=[
            {"type": "text", "text": initial_prompt},
            {
                "type": "image_url",
                "image_url": {"url": f"data:image/jpeg;base64,{b64}"}
            },
        ]
    )
    
    response = llm.invoke([message])
    result = json.loads(response.content)
    # load.append(result)
    print("result:", json.dumps(result, indent=4))
    
    stage2_prompt = f"""
    Extract text from the image.

    Use the polygon coordinates to know which region belongs to which polygon_id.
    Use the Stage-1 result only as a reference for polygon_id mapping.

    For each polygon_id:
    - Read the text inside that polygon region
    - Extract key-value pairs from the visible text
    - Return the extracted text as key-value dictionaries

    Stage-1 result (reference):
    {json.dumps(result, indent=2)}

    Polygon coordinates:
    {json.dumps(polygons, indent=2)}

    Return ONLY valid JSON.
    Do not include any explanation, text, or comments.

    OUTPUT FORMAT (STRICT JSON ONLY):
    {{
    "documents": [
        {{
        "polygon_id": <int>,
        "fields": {{
            "<key>": "<value>"
        }}
        }}
    ]
    }}
    """
    
    
    message_1 = HumanMessage(
        content=[
            {"type": "text", "text": stage2_prompt},
            {
                "type": "image_url",
                "image_url": {"url": f"data:image/jpeg;base64,{b64}"}
            },
        ]
    )

    final_response = llm.invoke([message_1])
    final_result = json.loads(final_response.content)
    load.append(final_result)
    print('Final:',json.dumps(final_result, indent=4))
````

````python
# Revise later - Ner + embedding 
from transformers import pipeline
from sentence_transformers import SentenceTransformer
import numpy as np

ner = pipeline("ner",model="dslim/bert-base-NER")

embedder = SentenceTransformer("all-MiniLM-L6-v2")


SCHEMA = {
    "company_name": "Name of the company or organization where a person works",
    "location": "City, state, or country where something is located",
    "person_name": "Full name of a person",
    "date": "Date or year"
}
def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# def extract_entities(text):
text = "Peter works at Amazon in Bangalore since 2022."

ner_results = ner(text)

schema_embeddings = {
    k: embedder.encode(v) for k, v in SCHEMA.items()
}

output = []

for ent in ner_results:
    value = ent["word"]
    ner_label = ent["entity"]

    value_embedding = embedder.encode(value)

    # Compare against schema
    scores = {}
    for field, emb in schema_embeddings.items():
        scores[field] = cosine_sim(value_embedding, emb)

    best_field = max(scores, key=scores.get)

    output.append({
        "value": value,
        "ner_label": ner_label,
        "mapped_field": best_field,
        "confidence": round(scores[best_field], 3)
    })

output
````

````python
# Document detection using countours + clustering : Stage -3(Less accuracy)
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

for image_path in paths:
    image = cv2.imread(image_path)
    canvas = image.copy()

    H, W = image.shape[:2]

    # -----------------------------
    # PREPROCESS
    # -----------------------------
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    edged = cv2.Canny(gray, 75, 200)

    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (15, 15))
    edged = cv2.morphologyEx(edged, cv2.MORPH_CLOSE, kernel)

    # -----------------------------
    # FIND CONTOURS
    # -----------------------------
    contours, _ = cv2.findContours(
        edged,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    # -----------------------------
    # FILTER CONTOURS (KEEP FRAGMENTS)
    # -----------------------------
    MIN_AREA = 0.005 * H * W

    valid = []
    for c in contours:
        if cv2.contourArea(c) > MIN_AREA:
            x, y, w, h = cv2.boundingRect(c)
            valid.append((x, y, w, h, c))

    if len(valid) == 0:
        continue

    # -----------------------------
    # SORT BY Y (STACKED DOC FIX)
    # -----------------------------
    valid = sorted(valid, key=lambda b: b[1])

    # -----------------------------
    # VERTICAL GROUPING
    # -----------------------------
    vertical_groups = []
    current_group = [valid[0]]

    Y_GAP_THRESH = 50   # tune: 30–60

    for prev, curr in zip(valid, valid[1:]):

        _, y1, _, h1, _ = prev
        _, y2, _, _, _ = curr

        if y2 - (y1 + h1) < Y_GAP_THRESH:
            current_group.append(curr)
        else:
            vertical_groups.append(current_group)
            current_group = [curr]

    vertical_groups.append(current_group)

    # -----------------------------
    # PROCESS EACH DOCUMENT
    # -----------------------------
    doc_id = 1

    for group in vertical_groups:

        contours_in_group = []
        centers = []

        for (_, _, _, _, c) in group:
            contours_in_group.append(c)

            M = cv2.moments(c)
            if M["m00"] != 0:
                cx = int(M["m10"] / M["m00"])
                cy = int(M["m01"] / M["m00"])
                centers.append([cx, cy])

        centers = np.array(centers)

        # -----------------------------
        # CLUSTER (SAFETY FOR MERGED BLOCKS)
        # -----------------------------
        if len(centers) > 3:
            labels = DBSCAN(eps=120, min_samples=2).fit(centers).labels_
        else:
            labels = np.zeros(len(contours_in_group))

        # -----------------------------
        # MERGE → CORNERS
        # -----------------------------
        for label in set(labels):

            if label == -1:
                continue

            idxs = np.where(labels == label)[0]

            merged = []
            for i in idxs:
                merged.append(contours_in_group[i])

            merged = np.vstack(merged)

            rect = cv2.minAreaRect(merged)
            box = cv2.boxPoints(rect)
            box = np.int32(box)

            # -----------------------------
            # ORDER CORNERS INLINE
            # -----------------------------
            pts = box.astype("float32")
            ordered = np.zeros((4, 2), dtype="float32")

            s = pts.sum(axis=1)
            ordered[0] = pts[np.argmin(s)]   # top-left
            ordered[2] = pts[np.argmax(s)]   # bottom-right

            diff = np.diff(pts, axis=1)
            ordered[1] = pts[np.argmin(diff)]  # top-right
            ordered[3] = pts[np.argmax(diff)]  # bottom-left

            # -----------------------------
            # DRAW RESULT
            # -----------------------------
            cv2.drawContours(canvas, [ordered.astype(int)], -1, (0, 255, 0), 3)

            for (x, y) in ordered:
                cv2.circle(canvas, (int(x), int(y)), 6, (0, 0, 255), -1)

            cv2.putText(
                canvas,
                f"Doc {doc_id}",
                (int(ordered[0][0]), int(ordered[0][1] - 10)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (255, 0, 0),
                2
            )

            doc_id += 1

    # -----------------------------
    # DISPLAY
    # -----------------------------
    plt.figure(figsize=(8, 12))
    plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
    plt.title("All Documents Detected")
    plt.axis("off")
    plt.show()
````

````python
# Detecting documnets using CV-contours(phase -3)
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
import os

files = [os.path.relpath(os.path.join(r, f), 'sample') for r, _, fs in os.walk('sample') for f in fs if f.lower().endswith(('png', 'jpg', 'jpeg'))]

for image_path in paths:

    image = cv2.imread(image_path)
    canvas = image.copy()

    H, W = image.shape[:2]

    # -----------------------------
    # PREPROCESS
    # -----------------------------
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    edged = cv2.Canny(gray, 75, 200)

    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (15, 15))
    edged = cv2.morphologyEx(edged, cv2.MORPH_CLOSE, kernel)

    # -----------------------------
    # FIND CONTOURS
    # -----------------------------
    contours, _ = cv2.findContours(
        edged,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    # -----------------------------
    # FILTER CONTOURS (KEEP FRAGMENTS)
    # -----------------------------
    MIN_AREA = 0.005 * H * W

    valid = []
    for c in contours:
        if cv2.contourArea(c) > MIN_AREA:
            x, y, w, h = cv2.boundingRect(c)
            valid.append((x, y, w, h, c))

    if len(valid) == 0:
        continue

    # -----------------------------
    # SORT BY Y (STACKED DOC FIX)
    # -----------------------------
    valid = sorted(valid, key=lambda b: b[1])

    # -----------------------------
    # VERTICAL GROUPING
    # -----------------------------
    vertical_groups = []
    current_group = [valid[0]]

    Y_GAP_THRESH = 50   # tune: 30–60

    for prev, curr in zip(valid, valid[1:]):

        _, y1, _, h1, _ = prev
        _, y2, _, _, _ = curr

        if y2 - (y1 + h1) < Y_GAP_THRESH:
            current_group.append(curr)
        else:
            vertical_groups.append(current_group)
            current_group = [curr]

    vertical_groups.append(current_group)

    # -----------------------------
    # PROCESS EACH DOCUMENT
    # -----------------------------
    doc_id = 1

    for group in vertical_groups:

        contours_in_group = []
        centers = []

        for (_, _, _, _, c) in group:
            contours_in_group.append(c)

            M = cv2.moments(c)
            if M["m00"] != 0:
                cx = int(M["m10"] / M["m00"])
                cy = int(M["m01"] / M["m00"])
                centers.append([cx, cy])

        centers = np.array(centers)

        # -----------------------------
        # CLUSTER (SAFETY FOR MERGED BLOCKS)
        # -----------------------------
        if len(centers) > 3:
            labels = DBSCAN(eps=120, min_samples=2).fit(centers).labels_
        else:
            labels = np.zeros(len(contours_in_group))

        # -----------------------------
        # MERGE → CORNERS
        # -----------------------------
        for label in set(labels):

            if label == -1:
                continue

            idxs = np.where(labels == label)[0]

            merged = []
            for i in idxs:
                merged.append(contours_in_group[i])

            merged = np.vstack(merged)

            rect = cv2.minAreaRect(merged)
            box = cv2.boxPoints(rect)
            box = np.int32(box)

            # -----------------------------
            # ORDER CORNERS INLINE
            # -----------------------------
            pts = box.astype("float32")
            ordered = np.zeros((4, 2), dtype="float32")

            s = pts.sum(axis=1)
            ordered[0] = pts[np.argmin(s)]   # top-left
            ordered[2] = pts[np.argmax(s)]   # bottom-right

            diff = np.diff(pts, axis=1)
            ordered[1] = pts[np.argmin(diff)]  # top-right
            ordered[3] = pts[np.argmax(diff)]  # bottom-left

            # -----------------------------
            # DRAW RESULT
            # -----------------------------
            cv2.drawContours(canvas, [ordered.astype(int)], -1, (0, 255, 0), 3)

            for (x, y) in ordered:
                cv2.circle(canvas, (int(x), int(y)), 6, (0, 0, 255), -1)

            cv2.putText(
                canvas,
                f"Doc {doc_id}",
                (int(ordered[0][0]), int(ordered[0][1] - 10)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (255, 0, 0),
                2
            )

            doc_id += 1

    # -----------------------------
    # DISPLAY
    # -----------------------------
    plt.figure(figsize=(8, 12))
    plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
    plt.title("All Documents Detected")
    plt.axis("off")
    plt.show()
````

````python
# Azure Right side value

def get_value_right_of_key(lines, key_text):
    # find the key line
    key_line = None
    for line in lines:
        if key_text.lower() in line['content'].lower():
            key_line = line
            break

    if not key_line:
        return None

    # key coordinates
    kx1, ky1, kx2, ky2, kx3, ky3, kx4, ky4 = key_line['polygon']
    key_y_top = min(ky1, ky2, ky3, ky4)
    key_y_bottom = max(ky1, ky2, ky3, ky4)
    key_x_right = max(kx1, kx2, kx3, kx4)

    candidates = []

    for line in lines:
        if line == key_line:
            continue

        px1, py1, px2, py2, px3, py3, px4, py4 = line['polygon']
        line_y_top = min(py1, py2, py3, py4)
        line_y_bottom = max(py1, py2, py3, py4)
        line_x_left = min(px1, px2, px3, px4)

        # Check horizontal alignment (same row)
        overlap = not (line_y_bottom < key_y_top or line_y_top > key_y_bottom)

        # Check if to the right
        if overlap and line_x_left > key_x_right:
            candidates.append((line_x_left, line['content']))

    # Pick nearest on right
    if candidates:
        candidates.sort()
        return candidates[0][1]

    return None


expiry = get_value_right_of_key(lines, "class")
print(expiry)
````

````python
# Azure Heuristic-based extraction
import re
from collections import OrderedDict

DATE_PATTERN = r"\d{4}/\d{2}/\d{2}|\d{2}/\d{2}/\d{4}"

KEY_SYNONYMS = {
    "date of birth": ["dob", "birth date", "d.o.b"],
    "expiry date": ["expiry", "valid until"],
    "issue date": ["issued", "first issued"],
    "nationality": ["nat."],
    "name": []
}

FIELD_PRIORITY = [
    "name",
    "nationality",
    "date of birth",
    "issue date",
    "expiry date"
]

NOISE_KEYWORDS = [
    "SULTANATE", "DIRECTORATE", "LICENCE"
]


def clean_empty(lines):
    return [line.strip() for line in lines if line.strip()]


def remove_duplicates(lines):
    return list(OrderedDict.fromkeys(lines))


def remove_noise(lines):
    return [
        line for line in lines
        if not any(word.lower() in line.lower() for word in NOISE_KEYWORDS)
    ]


def remove_numeric_noise(lines):
    cleaned = []
    for line in lines:
        if re.fullmatch(r"\d+", line) and not re.search(DATE_PATTERN, line):
            continue
        cleaned.append(line)
    return cleaned

def stitch_colon_lines(lines):
    stitched = []
    skip = False

    for i in range(len(lines)):
        if skip:
            skip = False
            continue

        if lines[i].endswith(":") and i + 1 < len(lines):
            stitched.append(f"{lines[i]} {lines[i+1]}")
            skip = True
        else:
            stitched.append(lines[i])

    return stitched


def merge_multiline_values(lines):
    merged = []
    for line in lines:
        if merged and ":" not in line:
            merged[-1] += " " + line
        else:
            merged.append(line)
    return merged

def attach_lonely_dates(lines):
    for i, line in enumerate(lines):
        if re.search(DATE_PATTERN, line) and ":" not in line:
            window = lines[max(0, i-2): i+3]
            for w in window:
                if ":" in w:
                    lines[i] = w + " " + line
                    break
    return lines

def handle_table_lines(lines):
    new_lines = []
    for line in lines:
        if "|" in line or "\t" in line:
            new_lines.append(" ".join(line.split()))
        else:
            new_lines.append(line)
    return new_lines

def normalize_keywords(text):
    for standard, variants in KEY_SYNONYMS.items():
        for v in variants:
            text = re.sub(rf"\b{re.escape(v)}\b", standard, text, flags=re.IGNORECASE)
    return text

def apply_field_priority(lines):
    prioritized = []

    for field in FIELD_PRIORITY:
        for line in lines:
            if field in line.lower():
                prioritized.append(line)

    for line in lines:
        if line not in prioritized:
            prioritized.append(line)

    return prioritized

def format_for_llm(lines):
    return "\n".join(f"- {line.strip()}" for line in lines)

def preprocess_ocr_lines(lines):
    lines = [s for s in lines if not re.search(r'[\u0600-\u06FF]', s)]
    lines = clean_empty(lines)
    lines = remove_duplicates(lines)
    lines = stitch_colon_lines(lines)
    lines = merge_multiline_values(lines)
    lines = attach_lonely_dates(lines)
    lines = remove_numeric_noise(lines)
    lines = remove_noise(lines)
    lines = handle_table_lines(lines)

    text = "\n".join(lines)
    text = normalize_keywords(text)

    lines = text.split("\n")
    lines = apply_field_priority(lines)

    return format_for_llm(lines)
````

````python
# Spatial_detection
import cv2
import dlib
from scipy.spatial import distance

# EAR calculation
def eye_aspect_ratio(eye):
    A = distance.euclidean(eye[1], eye[5])
    B = distance.euclidean(eye[2], eye[4])
    C = distance.euclidean(eye[0], eye[3])
    return (A + B) / (2.0 * C)

# Constants
EAR_THRESHOLD = 0.25
CONSEC_FRAMES = 3

blink_counter = 0
total_blinks = 0

# Load models
detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor("shape_predictor_68_face_landmarks.dat")

# Eye landmark indexes
LEFT_EYE = list(range(42, 48))
RIGHT_EYE = list(range(36, 42))

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = detector(gray)

    for face in faces:
        shape = predictor(gray, face)
        coords = [(shape.part(i).x, shape.part(i).y) for i in range(68)]

        left_eye = [coords[i] for i in LEFT_EYE]
        right_eye = [coords[i] for i in RIGHT_EYE]

        leftEAR = eye_aspect_ratio(left_eye)
        rightEAR = eye_aspect_ratio(right_eye)

        ear = (leftEAR + rightEAR) / 2.0

        if ear < EAR_THRESHOLD:
            blink_counter += 1
        else:
            if blink_counter >= CONSEC_FRAMES:
                total_blinks += 1
                print("Blink detected!")
            blink_counter = 0

        cv2.putText(frame, f"Blinks: {total_blinks}", (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

    cv2.imshow("Blink Detection", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
````

````python
# liveness detection using MediaPipe Face Mesh
import cv2
import mediapipe as mp
import numpy as np

mp_face_mesh = mp.solutions.face_mesh
face_mesh = mp_face_mesh.FaceMesh(refine_landmarks=True)

cap = cv2.VideoCapture(0)

blink_count = 0
prev_ear = None
head_turn_detected = False

def eye_aspect_ratio(landmarks, eye_indices, w, h):
    points = [(int(landmarks[i].x * w), int(landmarks[i].y * h)) for i in eye_indices]

    A = np.linalg.norm(np.array(points[1]) - np.array(points[5]))
    B = np.linalg.norm(np.array(points[2]) - np.array(points[4]))
    C = np.linalg.norm(np.array(points[0]) - np.array(points[3]))

    return (A + B) / (2.0 * C)

LEFT_EYE = [362, 385, 387, 263, 373, 380]
RIGHT_EYE = [33, 160, 158, 133, 153, 144]

while True:
    ret, frame = cap.read()
    h, w = frame.shape[:2]

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    result = face_mesh.process(rgb)

    status = "SPOOF ❌"

    if result.multi_face_landmarks:
        for face_landmarks in result.multi_face_landmarks:

            landmarks = face_landmarks.landmark

            # 👁️ Blink
            leftEAR = eye_aspect_ratio(landmarks, LEFT_EYE, w, h)
            rightEAR = eye_aspect_ratio(landmarks, RIGHT_EYE, w, h)
            ear = (leftEAR + rightEAR) / 2.0

            if prev_ear is not None:
                if ear < 0.22 and prev_ear > 0.25:
                    blink_count += 1
                    print("Blink!")

            prev_ear = ear

            # 🧠 Head turn (nose movement)
            nose_x = landmarks[1].x

            if nose_x < 0.4 or nose_x > 0.6:
                head_turn_detected = True

            # 🎯 Final logic
            if blink_count >= 2 and head_turn_detected:
                status = "LIVE ✅"

    cv2.putText(frame, f"Blinks: {blink_count}", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)

    cv2.putText(frame, status, (10, 60),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0,255,0), 2)

    cv2.imshow("Liveness", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
````

````python
# LayoutLMv3Processor
import os
import json
import torch
from PIL import Image
from transformers import LayoutLMv3Processor, LayoutLMv3ForTokenClassification

# =========================
# CONFIG
# =========================
DATASET_PATH = "dataset"
MODEL_PATH = "license_model"

FIELDS = ["NAME", "DOB", "NATIONALITY"]

# =========================
# GROUND TRUTH (YOU FILL THIS)
# =========================
GROUND_TRUTH_MAP = {
    "1.jpg": {
        "NAME": "Arushi Arvind Kumar Swarnkar",
        "DOB": "02-12-1998",
        "NATIONALITY": "India"
    },
    "2.jpg": {
        "NAME": "Lamiaa Imam Hassan Morsi",
        "DOB": "30-08-1970",
        "NATIONALITY": "Egypt"
    },
    "3.png": {
        "NAME": "Mohammed Akram Nahar Ashraa",
        "DOB": "28-12-1982",
        "NATIONALITY": "Jordan"
    },
    "4.png": {
        "NAME": "Abdurakhim Diya",
        "DOB": "20-04-1994",
        "NATIONALITY": "Kazakhstan"
    },
    "5.png": {
        "NAME": "Abdulkarim Yousef A Khamas",
        "DOB": "30-12-1956",
        "NATIONALITY": "Emirates"
    }
}

# =========================
# LABEL SETUP
# =========================
label_list = ["O"]
for field in FIELDS:
    label_list.append(f"B-{field}")
    label_list.append(f"I-{field}")

label_map = {l: i for i, l in enumerate(label_list)}

# =========================
# HELPERS
# =========================
def polygon_to_bbox(polygon):
    x = polygon[0::2]
    y = polygon[1::2]
    return [min(x), min(y), max(x), max(y)]

def extract_words_boxes(data):
    words, boxes = [], []
    for item in data:
        text = item["content"].strip()
        if text == "":
            continue
        words.append(text)
        boxes.append(polygon_to_bbox(item["polygon"]))
    return words, boxes

def split_words(words, boxes):
    new_words, new_boxes = [], []
    for w, b in zip(words, boxes):
        for p in w.split():
            new_words.append(p)
            new_boxes.append(b)
    return new_words, new_boxes

def normalize_boxes(boxes, width, height):
    return [
        [
            int(1000 * b[0] / width),
            int(1000 * b[1] / height),
            int(1000 * b[2] / width),
            int(1000 * b[3] / height),
        ]
        for b in boxes
    ]

def create_labels(words, ground_truth):
    labels = ["O"] * len(words)

    for field, value in ground_truth.items():
        value_words = value.split()

        for i in range(len(words)):
            if words[i:i+len(value_words)] == value_words:

                labels[i] = f"B-{field}"

                for j in range(1, len(value_words)):
                    if i + j < len(words):
                        labels[i+j] = f"I-{field}"

    return labels

# =========================
# LOAD MODEL
# =========================
processor = LayoutLMv3Processor.from_pretrained(
    "microsoft/layoutlmv3-base",
    apply_ocr=False
)

model = LayoutLMv3ForTokenClassification.from_pretrained(
    "microsoft/layoutlmv3-base",
    num_labels=len(label_list)
)

# =========================
# BUILD DATASET
# =========================
dataset = []

for file in os.listdir(DATASET_PATH):

    if file.endswith(".png") or file.endswith(".jpg"):

        if file not in GROUND_TRUTH_MAP:
            print(f"Skipping {file} (no ground truth)")
            continue

        image_path = os.path.join(DATASET_PATH, file)
        json_path = image_path.replace(".png", ".json").replace(".jpg", ".json")

        if not os.path.exists(json_path):
            print(f"Missing JSON for {file}")
            continue

        image = Image.open(image_path).convert("RGB")
        width, height = image.size

        with open(json_path, "r", encoding="utf-8") as f:
            ocr_data = json.load(f)

        ground_truth = GROUND_TRUTH_MAP[file]

        words, boxes = extract_words_boxes(ocr_data)
        words, boxes = split_words(words, boxes)
        boxes = normalize_boxes(boxes, width, height)

        labels = create_labels(words, ground_truth)
        label_ids = [label_map[l] for l in labels]

        encoding = processor(
            image,
            words,
            boxes=boxes,
            word_labels=label_ids,
            return_tensors="pt",
            truncation=True,
            padding="max_length"
        )

        dataset.append(encoding)

print("Total documents loaded:", len(dataset))

# =========================
# TRAIN
# =========================
optimizer = torch.optim.AdamW(model.parameters(), lr=5e-5)

model.train()

for epoch in range(1):

    total_loss = 0

    for batch in dataset:
        outputs = model(**batch)
        loss = outputs.loss

        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

        total_loss += loss.item()

    print(f"Epoch {epoch} Loss:", total_loss)

# =========================
# SAVE MODEL
# =========================
model.save_pretrained(MODEL_PATH)
processor.save_pretrained(MODEL_PATH)

print("Model saved successfully!")

# =========================
# PREDICTION FUNCTION
# =========================
def predict(image_path, ocr_json):

    image = Image.open(image_path).convert("RGB")
    width, height = image.size

    words, boxes = extract_words_boxes(ocr_json)
    words, boxes = split_words(words, boxes)
    boxes = normalize_boxes(boxes, width, height)

    encoding = processor(
        image,
        words,
        boxes=boxes,
        return_tensors="pt",
        truncation=True,
        padding="max_length"
    )

    model.eval()
    with torch.no_grad():
        outputs = model(**encoding)

    preds = outputs.logits.argmax(-1).squeeze().tolist()
    word_ids = encoding.word_ids()

    results = {}
    current_field = None

    for idx, word_id in enumerate(word_ids):

        if word_id is None:
            continue

        label = label_list[preds[idx]]
        word = words[word_id]

        if label.startswith("B-"):
            current_field = label[2:]
            results[current_field] = word

        elif label.startswith("I-") and current_field:
            results[current_field] += " " + word

    return results
import json

test_file = r"C:\Users\peter.leo\Downloads\Projects\OCR\Custom model training_02_04\dataset\1.jpg"
json_path = r"C:\Users\peter.leo\Downloads\Projects\OCR\Custom model training_02_04\dataset\1.json"

# ✅ LOAD JSON PROPERLY
with open(json_path, "r", encoding="utf-8") as f:
    new_json = json.load(f)

# ✅ NOW PASS DATA (not path)
result = predict(test_file, new_json)

print(result)
````

````python
# Spacy nationality(Joint_venture) model
import random
import json

# Generate variations of "Nationality" with OCR errors
def generate_nationality_variations():
    """Generate all possible variations of 'Nationality' with OCR errors"""
    
    base = "Joint venture"
    
    # Common OCR substitutions
    substitutions = {
        'a': ['e', 'o', '@'],
        'i': ['l', '1', '!'],
        'o': ['0', 'O'],
        'l': ['1', 'I'],
        't': ['l', '1'],
        'n': ['m', 'r'],
        'e': ['a', 'c'],
        'y': ['j', 'v']
    }
    
    variations = set()
    variations.add(base)
    
    # 1. Single character substitutions
    for i in range(len(base)):
        char = base[i]
        if char in substitutions:
            for sub in substitutions[char]:
                variant = base[:i] + sub + base[i+1:]
                variations.add(variant)
    
    # 2. Missing characters (drop one)
    for i in range(len(base)):
        variant = base[:i] + base[i+1:]
        variations.add(variant)
    
    # 3. Double characters
    for i in range(len(base)-1):
        variant = base[:i+1] + base[i] + base[i+1:]
        variations.add(variant)
    return list(variations)

# Generate all variations
nationality_variations = generate_nationality_variations()

print(f"Total variations generated: {len(nationality_variations)}")
print("Sample variations:")
for var in nationality_variations[:20]:
    print(f"  {var}")

# Generate training data - just the word itself, no country
TRAIN_DATA = []
num_samples = 100

for _ in range(num_samples):
    # Pick a random variation of "Nationality"
    nationality_word = random.choice(nationality_variations)
    
    # Just the word itself, no extra text
    text = nationality_word
    
    # Entity spans the entire word
    start_idx = 0
    end_idx = len(nationality_word)
    
    # Add to training data
    TRAIN_DATA.append((text, {"entities": [(start_idx, end_idx, "Joint venture")]}))


# Print samples
print("\n" + "="*50)
print("Training Data Samples:")
print("="*50)

for i, (text, entities) in enumerate(TRAIN_DATA[:30]):
    print(f"{i+1}. '{text}' -> {entities}")

# Save to JSON
def save_to_json(train_data, filename="Joint_venture_training.json"):
    """Save training data as JSON"""
    json_data = []
    for text, annotations in train_data:
        json_data.append({
            "text": text,
            "entities": annotations["entities"]
        })
    
    with open(filename, 'w') as f:
        json.dump(json_data, f, indent=2)
    print(f"\nSaved {len(train_data)} examples to {filename}")

# Save to spaCy format
def save_to_spacy(train_data, filename="Joint_venture_training.spacy"):
    """Convert to spaCy DocBin format"""
    import spacy
    from spacy.tokens import DocBin
    
    nlp = spacy.blank("en")
    doc_bin = DocBin()
    
    for text, annotations in train_data:
        doc = nlp.make_doc(text)
        ents = []
        
        for start, end, label in annotations["entities"]:
            span = doc.char_span(start, end, label=label)
            if span is not None:
                ents.append(span)
        
        doc.ents = ents
        doc_bin.add(doc)
    
    doc_bin.to_disk(filename)
    print(f"Saved {len(train_data)} examples to {filename}")

# Save both formats
save_to_json(TRAIN_DATA)
save_to_spacy(TRAIN_DATA)

print("\n" + "="*50)
print("All variations of 'Joint_venture' used:")
print("="*50)
for var in sorted(nationality_variations):
    print(var)
````

````python
# Spacy nationality_model
import random
import spacy
from spacy.tokens import DocBin
from spacy.training import Example

def train_nationality_model(word):
    """Train a model to recognize variations of the given word"""
    
    # Generate variations of the word
    def generate_variations(base_word):
        substitutions = {
            'a': ['e', 'o', '@'],
            'i': ['l', '1', '!'],
            'o': ['0', 'O'],
            'l': ['1', 'I'],
            't': ['l', '1'],
            'n': ['m', 'r'],
            'e': ['a', 'c'],
            'y': ['j', 'v']
        }
        
        variations = set()
        variations.add(base_word)
        
        # Single character substitutions
        for i in range(len(base_word)):
            char = base_word[i]
            if char in substitutions:
                for sub in substitutions[char]:
                    variant = base_word[:i] + sub + base_word[i+1:]
                    variations.add(variant)
        
        # Missing characters
        for i in range(len(base_word)):
            variant = base_word[:i] + base_word[i+1:]
            variations.add(variant)
        
        # Double characters
        for i in range(len(base_word)-1):
            variant = base_word[:i+1] + base_word[i] + base_word[i+1:]
            variations.add(variant)
        
        return list(variations)
    
    # Generate all variations
    print(f"Generating variations for '{word}'...")
    variations = generate_variations(word)
    print(f"Found {len(variations)} variations")
    
    # Create training data - ALL variations are labeled as the ORIGINAL word
    train_data = []
    for variant in variations:
        for _ in range(3):
            # The text is the misspelled version, but the entity is the original word
            train_data.append((variant, {"entities": [(0, len(variant), word)]}))
    
    # Convert to spaCy format
    nlp = spacy.blank("en")
    doc_bin = DocBin()
    
    for text, annotations in train_data:
        doc = nlp.make_doc(text)
        ents = []
        for start, end, label in annotations["entities"]:
            span = doc.char_span(start, end, label=label)
            if span is not None:
                ents.append(span)
        doc.ents = ents
        doc_bin.add(doc)
    
    doc_bin.to_disk("training_data.spacy")
    print(f"Created {len(train_data)} training examples")
    
    # Train the model
    print("\nTraining model...")
    nlp = spacy.blank("en")
    ner = nlp.add_pipe("ner")
    ner.add_label(word)  # Label is the original word
    
    # Load training data
    doc_bin = DocBin().from_disk("training_data.spacy")
    train_docs = list(doc_bin.get_docs(nlp.vocab))
    
    examples = []
    for doc in train_docs:
        examples.append(Example.from_dict(doc, {
            "entities": [(ent.start_char, ent.end_char, ent.label_) for ent in doc.ents]
        }))
    
    # Train
    nlp.begin_training()
    for epoch in range(30):
        random.shuffle(examples)
        losses = {}
        for batch in spacy.util.minibatch(examples, size=16):
            nlp.update(batch, losses=losses)
        
        if epoch % 10 == 0:
            print(f"Epoch {epoch}, Loss: {losses.get('ner', 0):.4f}")
    
    # Save model
    nlp.to_disk("trained_model")
    print(f"\nModel saved to 'trained_model'")
    
    return nlp

# Test the trained model
def test_model(original_word):
    """Test the trained model"""
    nlp = spacy.load("trained_model")
    
    # Test with various misspellings
    test_words = [
        original_word,
        "Natinality",
        "Nationlity",
        "Nationalty",
        "Nati0nality",
        "Nati0naliity",
        "Natlonality",
        "Natiionality",
        "Nati0nal1ty",
        "Natonality",
        "Nationallity",
        "Nat1onality",
        "Nati0naltiy",
        "Nationaiity",
        "N@tionality",
        "NationaIity",
        "Natinalty",
        "Nationlty"
    ]
    
    print("\n" + "="*50)
    print("Testing the Model:")
    print("="*50)
    
    for text in test_words:
        doc = nlp(text)
        if doc.ents:
            for ent in doc.ents:
                print(f"✓ '{text}' -> Recognized as: '{ent.label_}' (original: {original_word})")
        else:
            print(f"✗ '{text}' -> Not recognized")
    
    # Test with longer text
    print("\n" + "="*50)
    print("Testing with full sentences:")
    print("="*50)
    
    sentences = [
        f"Please enter your Natinality",
        f"Nationlity: Required field",
        f"Your Nati0nality is important",
        f"Select your Nationalty from the list"
    ]
    
    for sentence in sentences:
        doc = nlp(sentence)
        print(f"\nText: {sentence}")
        for ent in doc.ents:
            print(f"  Entity: '{ent.text}' -> Recognized as '{ent.label_}'")

# Simple usage
if __name__ == "__main__":
    # Train with any word you want
    word_to_train = "Nationality"  # Change this to any word
    
    model = train_nationality_model(word_to_train)
    test_model(word_to_train)
    
    print("\n" + "="*50)
    print("How to use the trained model:")
    print("="*50)
    print("""
import spacy
nlp = spacy.load("trained_model")
text = "Natinality"
doc = nlp(text)
for ent in doc.ents:
    print(f"Found: {ent.text} -> {ent.label_}")
# Output: Found: Natinality -> Nationality
    """)
````

````python
# Spelling Correction using SymSpell
from symspellpy import SymSpell

class SpellCorrector:
    def __init__(self, dict_path):
        self.sym_spell = SymSpell(max_dictionary_edit_distance=2, prefix_length=7)
        self.sym_spell.load_dictionary(dict_path, 0, 1)

    def correct(self, text):
        if len(text) > 4:
            suggestions = self.sym_spell.lookup_compound(text, max_edit_distance=2)
            return suggestions[0].term if suggestions else text
        return text

def build_dictionary(base_file, custom_file, output_file):
    word_freq = {}
    with open(base_file, "r", encoding="utf-8-sig") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 2: continue
            word, freq = parts
            word_freq[word] = int(freq)
    with open(custom_file, "r", encoding="utf-8-sig") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 2: continue
            word, freq = parts
            word_freq[word] = int(freq)
    with open(output_file, "w", encoding="utf-8-sig") as f:
        for word, freq in word_freq.items(): f.write(f"{word} {freq}\n")
prebuild_file = 'frequency_dictionary_en_82_765.txt'
custom_file = 'custom_dict.txt'
final_file = 'final_dict.txt'

build_dictionary(prebuild_file, custom_file, final_file)
model = SpellCorrector('final_dict.txt')
spell_check = model.correct("illuson")
spell_check
````

````python
# Nationality spacy model
import random
import spacy
from spacy.tokens import DocBin
from spacy.training import Example

def train_nationality_model(word):
    """Train a model to recognize variations of the given word"""
    
    # Generate variations of the word
    def generate_variations(base_word):
        substitutions = {
            'a': ['e', 'o', '@'],
            'i': ['l', '1', '!'],
            'o': ['0', 'O'],
            'l': ['1', 'I'],
            't': ['l', '1'],
            'n': ['m', 'r'],
            'e': ['a', 'c'],
            'y': ['j', 'v']
        }
        
        variations = set()
        variations.add(base_word)
        
        # Single character substitutions
        for i in range(len(base_word)):
            char = base_word[i]
            if char in substitutions:
                for sub in substitutions[char]:
                    variant = base_word[:i] + sub + base_word[i+1:]
                    variations.add(variant)
        
        # Missing characters
        for i in range(len(base_word)):
            variant = base_word[:i] + base_word[i+1:]
            variations.add(variant)
        
        # Double characters
        for i in range(len(base_word)-1):
            variant = base_word[:i+1] + base_word[i] + base_word[i+1:]
            variations.add(variant)
        
        return list(variations)
    
    # Generate all variations
    print(f"Generating variations for '{word}'...")
    variations = generate_variations(word)
    print(f"Found {len(variations)} variations")
    
    # Create training data - ALL variations are labeled as the ORIGINAL word
    train_data = []
    for variant in variations:
        for _ in range(3):
            # The text is the misspelled version, but the entity is the original word
            train_data.append((variant, {"entities": [(0, len(variant), word)]}))
    
    # Convert to spaCy format
    nlp = spacy.blank("en")
    doc_bin = DocBin()
    
    for text, annotations in train_data:
        doc = nlp.make_doc(text)
        ents = []
        for start, end, label in annotations["entities"]:
            span = doc.char_span(start, end, label=label)
            if span is not None:
                ents.append(span)
        doc.ents = ents
        doc_bin.add(doc)
    
    doc_bin.to_disk("training_data.spacy")
    print(f"Created {len(train_data)} training examples")
    
    # Train the model
    print("\nTraining model...")
    nlp = spacy.blank("en")
    ner = nlp.add_pipe("ner")
    ner.add_label(word)  # Label is the original word
    
    # Load training data
    doc_bin = DocBin().from_disk("training_data.spacy")
    train_docs = list(doc_bin.get_docs(nlp.vocab))
    
    examples = []
    for doc in train_docs:
        examples.append(Example.from_dict(doc, {
            "entities": [(ent.start_char, ent.end_char, ent.label_) for ent in doc.ents]
        }))
    
    # Train
    nlp.begin_training()
    for epoch in range(30):
        random.shuffle(examples)
        losses = {}
        for batch in spacy.util.minibatch(examples, size=16):
            nlp.update(batch, losses=losses)
        
        if epoch % 10 == 0:
            print(f"Epoch {epoch}, Loss: {losses.get('ner', 0):.4f}")
    
    # Save model
    nlp.to_disk("trained_model")
    print(f"\nModel saved to 'trained_model'")
    
    return nlp

# Test the trained model
def test_model(original_word):
    """Test the trained model"""
    nlp = spacy.load("trained_model")
    
    # Test with various misspellings
    test_words = [
        original_word,
        "Natinality",
        "Nationlity",
        "Nationalty",
        "Nati0nality",
        "Nati0naliity",
        "Natlonality",
        "Natiionality",
        "Nati0nal1ty",
        "Natonality",
        "Nationallity",
        "Nat1onality",
        "Nati0naltiy",
        "Nationaiity",
        "N@tionality",
        "NationaIity",
        "Natinalty",
        "Nationlty"
    ]
    
    print("\n" + "="*50)
    print("Testing the Model:")
    print("="*50)
    
    for text in test_words:
        doc = nlp(text)
        if doc.ents:
            for ent in doc.ents:
                print(f"✓ '{text}' -> Recognized as: '{ent.label_}' (original: {original_word})")
        else:
            print(f"✗ '{text}' -> Not recognized")
    
    # Test with longer text
    print("\n" + "="*50)
    print("Testing with full sentences:")
    print("="*50)
    
    sentences = [
        f"Please enter your Natinality",
        f"Nationlity: Required field",
        f"Your Nati0nality is important",
        f"Select your Nationalty from the list"
    ]
    
    for sentence in sentences:
        doc = nlp(sentence)
        print(f"\nText: {sentence}")
        for ent in doc.ents:
            print(f"  Entity: '{ent.text}' -> Recognized as '{ent.label_}'")

# Simple usage
if __name__ == "__main__":
    # Train with any word you want
    word_to_train = "Nationality"  # Change this to any word
    
    model = train_nationality_model(word_to_train)
    test_model(word_to_train)
    
    print("\n" + "="*50)
    print("How to use the trained model:")
    print("="*50)
    print("""
import spacy
nlp = spacy.load("trained_model")
text = "Natinality"
doc = nlp(text)
for ent in doc.ents:
    print(f"Found: {ent.text} -> {ent.label_}")
# Output: Found: Natinality -> Nationality
    """)
````

````python
# KYC Document Validation System using FastAPI (19-03-2026)

from fastapi import FastAPI, File, UploadFile, Form, Request
from fastapi.responses import HTMLResponse, JSONResponse, FileResponse
from fastapi.templating import Jinja2Templates
import shutil, tempfile, os, uuid, cv2, numpy as np
from pathlib import Path
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning)
warnings.filterwarnings('ignore', category=FutureWarning)

# from utils import (ocr_extract, keyword_match, doc_expiry, quality_check, similarity_search)

app = FastAPI(title="Document Validation System")

UPLOAD_DIR = Path("uploads")
UPLOAD_DIR.mkdir(exist_ok=True)
templates = Jinja2Templates(directory="templates")

def extract_text_from_region(image_array, x, y, w, h):
    try:
        cropped = image_array[y:y+h, x:x+w]
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp:
            cv2.imwrite(tmp.name, cropped)
            tmp_path = tmp.name
        try:
            extracted_texts = ocr_extract(tmp_path)
            return extracted_texts
        finally: os.unlink(tmp_path)
    except Exception as e:
        print(f"Region extraction error: {e}")
        return []

@app.get("/", response_class=HTMLResponse)
async def home(request: Request):
    """Serve the main HTML page"""
    return templates.TemplateResponse("index.html", {"request": request})

@app.post("/upload")
async def upload_file(file: UploadFile = File(...)):
    """Simple upload endpoint - just saves file"""
    try:
        filename = f"{uuid.uuid4()}{Path(file.filename).suffix}"
        filepath = UPLOAD_DIR / filename
        with open(filepath, "wb") as buffer:
            shutil.copyfileobj(file.file, buffer)
        return {"success": True, "filename": filename}
    except Exception as e: return {"success": False, "error": str(e)}

@app.post("/validate")
async def validate_document(
    validation_type: str = Form(...), keywords: str = Form(""),
    image1: UploadFile = File(...), image2: UploadFile = File(None),
    region_x: int = Form(0), region_y: int = Form(0),
    region_width: int = Form(0), region_height: int = Form(0)):
    try:
        if validation_type == "face_similarity":
            if image2 is None: return {"error": "Second image required for face similarity"}
            result = await similarity_search(image1, image2)
            return result
        image1_bytes = await image1.read()
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp1:
            tmp1.write(image1_bytes)
            img1_path = tmp1.name
        try:
            if validation_type == "region_selection":
                if region_width > 0 and region_height > 0:
                    img = cv2.imread(img1_path)
                    extracted_texts = extract_text_from_region(img, region_x, region_y, region_width, region_height)
                    full_text = " ".join(extracted_texts)
                    return {"success": True,"type": "region_selection","text": full_text,"word_count": len(extracted_texts)}
                else: return {"error": "Invalid region selected"}
            
            raw_text = ocr_extract(img1_path)
            if validation_type == "keyword_match":
                keyword_list = [k.strip() for k in keywords.split(",") if k.strip()]
                if not keyword_list: return {"error": "No keywords provided"}
                result = keyword_match(raw_text, keyword_list)
                return result
            elif validation_type == "expiration":
                result = doc_expiry(raw_text)
                return {"success": True,"type": "expiration","status": result}
                    # "all_text": " ".join(raw_text)[:500] + "..." if len(" ".join(raw_text)) > 500 else " ".join(raw_text)
            elif validation_type == "resolution":
                img = cv2.imread(img1_path)
                height, width = img.shape[:2]
                dimensions = [height, width, 3]
                result = quality_check(dimensions)
                return {"success": True,"type": "resolution","quality": result,"resolution": f"{width} x {height}"}
            else: return {"error": f"Unknown validation type: {validation_type}"}
        finally: os.unlink(img1_path)
    except Exception as e: return {"error": str(e)}

@app.get("/image/{filename}")
async def get_image(filename: str):
    """Serve uploaded images"""
    filepath = UPLOAD_DIR / filename
    if filepath.exists(): return FileResponse(filepath)
    return {"error": "Image not found"}

#############
# utils
#############

import re
from rapidfuzz import process, fuzz
from rapidocr_onnxruntime import RapidOCR
import cv2, os, tempfile
from datetime import datetime
from deepface import DeepFace

ocr = RapidOCR()

def fuzzy(fields,ocr_text,threshold=70):
    corrected_list = ocr_text.copy()
    for field in fields:
        match = process.extractOne(field,corrected_list,scorer=fuzz.partial_ratio)
        if match:
            matched_word, score, index = match
            print(match)
            if score >= threshold:
                corrected_list[index] = field
    return corrected_list

def normalize(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9 ]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def ocr_extract(image_path):
    result, _ = ocr(image_path)
    return [line[1] for line in result]

def keyword_match(raw_text,fields):
    norm_raw = [normalize(i) for i in raw_text]
    fuzzy_words = fuzzy(fields,norm_raw)
    score = 0
    for field in fields:
        if field in fuzzy_words:
            print(field)
            score += 1
    perc = (score/len(fields))*100
    return {'Score':score,'percentage':perc,'Matched':perc >= 65}

def doc_expiry(raw_text):
    dates = []
    for text in raw_text:
        matches = re.findall(r'\b\d{2}[-/]?\d{2}[-/]?\d{4}\b', text)
        for m in matches:
            clean = re.sub(r'[-/]', '', m)
            try:
                dt = datetime.strptime(clean,"%d%m%Y")
                dates.append(dt)
            except: pass

    if not dates: return 'No expiry date found'
    max_date, today = max(dates), datetime.today()
    return 'Valid' if max_date > today else 'Expired'

def quality_check(dim):
    height, width, _ = dim[0], dim[1], dim[2]
    print(dim)
    if int(height) < 640 or int(width) < 480: return 'Low Resolution Image'
    else: return 'Good Quality Image'
    
async def similarity_search(image1, image2):
    try:
        await image1.seek(0)
        await image2.seek(0)
        img1_bytes = await image1.read()
        img2_bytes = await image2.read()
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp1:
            tmp1.write(img1_bytes)
            tmp1_path = tmp1.name
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp2:
            tmp2.write(img2_bytes)
            tmp2_path = tmp2.name
        try:
            response = DeepFace.verify(img1_path=tmp1_path,img2_path=tmp2_path,model_name="ArcFace",
                detector_backend="retinaface",enforce_detection=False,align=True)
            return {"success": True,"type": "Face similarity","verified": response.get("verified", False),
                "confidence": response.get("confidence", 0),"time": response.get("time", 0)}
            # "model": "ArcFace",
            # "message": "Face similarity check completed"
            # "distance": response.get("distance", 0),
        except Exception as e: return {"success": False, "error": f"DeepFace verification failed: {str(e)}"}
        finally:
            try:
                os.unlink(tmp1_path)
                os.unlink(tmp2_path)
            except: pass
            
    except Exception as e: return {"success": False,"error": f"Face similarity failed: {str(e)}"}
    
############
# index.html
############
index_html = """
<!DOCTYPE html>
<html>
<head>
    <title>Document Validation System</title>
    <style>
        * {
            margin: 0;
            padding: 0;
            box-sizing: border-box;
            font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
        }
        
        body {
            background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
            min-height: 100vh;
            padding: 20px;
        }
        
        .container {
            max-width: 1200px;
            margin: 0 auto;
            background: white;
            border-radius: 15px;
            padding: 30px;
            box-shadow: 0 20px 60px rgba(0,0,0,0.3);
        }
        
        h1 {
            color: #333;
            margin-bottom: 10px;
        }
        
        .subtitle {
            color: #666;
            margin-bottom: 30px;
        }
        
        .flex-row {
            display: flex;
            gap: 20px;
            margin-bottom: 20px;
            flex-wrap: wrap;
        }
        
        .flex-col {
            flex: 1;
            min-width: 300px;
        }
        
        .card {
            background: #f8f9fa;
            padding: 20px;
            border-radius: 10px;
            border: 1px solid #e0e0e0;
            margin-bottom: 20px;
        }
        
        label {
            display: block;
            margin-bottom: 8px;
            font-weight: 600;
            color: #555;
        }
        
        select, input[type="file"], input[type="text"], textarea {
            width: 100%;
            padding: 12px;
            border: 2px solid #e0e0e0;
            border-radius: 8px;
            font-size: 16px;
            margin-bottom: 15px;
        }
        
        select:focus, input:focus, textarea:focus {
            outline: none;
            border-color: #667eea;
        }
        
        button {
            background: #667eea;
            color: white;
            border: none;
            padding: 12px 30px;
            border-radius: 8px;
            font-size: 16px;
            cursor: pointer;
            transition: background 0.3s;
            margin-right: 10px;
            margin-bottom: 10px;
        }
        
        button:hover {
            background: #5a67d8;
        }
        
        button.secondary {
            background: #48bb78;
        }
        
        button.secondary:hover {
            background: #38a169;
        }
        
        button:disabled {
            background: #ccc;
            cursor: not-allowed;
        }
        
        .image-container {
            background: #e9ecef;
            border-radius: 10px;
            overflow: hidden;
            margin: 20px 0;
            position: relative;
            border: 2px solid #ddd;
        }
        
        #documentImage {
            width: 100%;
            height: auto;
            display: block;
        }
        
        #selectionCanvas {
            position: absolute;
            top: 0;
            left: 0;
            width: 100%;
            height: 100%;
            cursor: crosshair;
            pointer-events: none;
        }
        
        #selectionCanvas.active {
            pointer-events: all;
        }
        
        .region-info {
            background: #e9ecef;
            padding: 10px;
            border-radius: 8px;
            font-family: monospace;
            margin: 10px 0;
        }
        
        .result-box {
            background: #f0f0f0;
            padding: 15px;
            border-radius: 8px;
            font-family: monospace;
            white-space: pre-wrap;
            max-height: 400px;
            overflow-y: auto;
            border: 1px solid #ddd;
            margin-top: 20px;
        }
        
        .preview-image {
            max-width: 100%;
            max-height: 200px;
            margin-top: 10px;
            border-radius: 8px;
            border: 1px solid #ddd;
            display: none;
        }
        
        .toast {
            position: fixed;
            bottom: 30px;
            right: 30px;
            background: #333;
            color: white;
            padding: 12px 24px;
            border-radius: 8px;
            display: none;
            z-index: 1000;
        }
        
        .toast.show {
            display: block;
        }
        
        .info-text {
            color: #666;
            font-size: 13px;
            margin-top: 5px;
        }
    </style>
</head>
<body>
    <div class="container">
        <h1>📄 KYC Document Validation System</h1>
        <div class="subtitle">Using RapidOCR + DeepFace for comprehensive validation</div>
        
        <!-- Step 1: Validation Type -->
        <div class="card">
            <label>1. Select Validation Type:</label>
            <select id="validationType">
                <option value="">-- Choose validation method --</option>
                <option value="keyword_match">🔍 Keyword Match</option>
                <option value="expiration">📅 Expiration Validation</option>
                <option value="resolution">📏 Resolution Check</option>
                <option value="face_similarity">👤 Face Similarity</option>
                <option value="region_selection">✂️ Select Document Region</option>
            </select>
        </div>
        
        <!-- Step 2: Upload Images -->
        <div class="flex-row">
            <div class="flex-col">
                <div class="card">
                    <label>2. Upload Document Image:</label>
                    <input type="file" id="docImage" accept="image/*">
                    <div class="image-container" id="docPreview" style="display: none;">
                        <img id="documentImage" src="" alt="Document">
                        <canvas id="selectionCanvas"></canvas>
                    </div>
                </div>
            </div>
            
            <div class="flex-col" id="selfieSection" style="display: none;">
                <div class="card">
                    <label>Upload Selfie Image (for face similarity):</label>
                    <input type="file" id="selfieImage" accept="image/*">
                    <img id="selfiePreview" class="preview-image" src="">
                </div>
            </div>
        </div>
        
        <!-- Step 3: Keywords (for keyword match) -->
        <div class="card" id="keywordSection" style="display: none;">
            <label>3. Enter Keywords (comma separated):</label>
            <input type="text" id="keywords" placeholder="e.g. name, date, passport, address">
            <div class="info-text">Keywords will be matched using fuzzy search</div>
        </div>
        
        <!-- Region Info (for region selection) -->
        <div id="regionInfo" class="region-info" style="display: none;">
            <span>✂️ Drag on image to select a region - Text will auto-copy!</span>
        </div>
        
        <!-- Selected Region Display -->
        <div id="selectedRegionDisplay" style="display: none; margin: 10px 0; padding: 10px; background: #e3f2fd; border-radius: 8px;">
            <strong>Selected Region:</strong> <span id="regionCoords"></span>
        </div>
        
        <!-- Validate Button -->
        <button id="validateBtn" class="secondary" style="margin-top: 10px;" disabled>🚀 Run Validation</button>
        
        <!-- Results -->
        <div class="card">
            <h3>Results:</h3>
            <div id="resultBox" class="result-box">
                Waiting for validation...
            </div>
        </div>
    </div>
    
    <div class="toast" id="toast"></div>
    
    <script>
        // ============ GLOBAL VARIABLES ============
        let docFilename = null;
        let selfieFilename = null;
        let selectedRegion = { x: 0, y: 0, width: 0, height: 0 };
        let isSelecting = false;
        let startX, startY;
        let scaleX = 1, scaleY = 1;  // Fixed: Added scale variables
        
        // DOM Elements
        const validationType = document.getElementById('validationType');
        const docImage = document.getElementById('docImage');
        const selfieImage = document.getElementById('selfieImage');
        const documentImg = document.getElementById('documentImage');
        const canvas = document.getElementById('selectionCanvas');
        const ctx = canvas.getContext('2d');
        const docPreview = document.getElementById('docPreview');
        const selfiePreview = document.getElementById('selfiePreview');
        const selfieSection = document.getElementById('selfieSection');
        const keywordSection = document.getElementById('keywordSection');
        const regionInfo = document.getElementById('regionInfo');
        const selectedRegionDisplay = document.getElementById('selectedRegionDisplay');
        const regionCoords = document.getElementById('regionCoords');
        const validateBtn = document.getElementById('validateBtn');
        const resultBox = document.getElementById('resultBox');
        const toast = document.getElementById('toast');
        
        // ============ EVENT LISTENERS ============
        
        // Validation type change
        validationType.addEventListener('change', function() {
            const type = this.value;
            
            // Show/hide sections based on type
            selfieSection.style.display = type === 'face_similarity' ? 'block' : 'none';
            keywordSection.style.display = type === 'keyword_match' ? 'block' : 'none';
            
            if (type === 'region_selection') {
                regionInfo.style.display = 'block';
                regionInfo.innerHTML = '✂️ Drag on image to select a region - Text will auto-copy!';
                canvas.classList.add('active');
            } else {
                regionInfo.style.display = 'none';
                canvas.classList.remove('active');
                selectedRegionDisplay.style.display = 'none';
                // Clear canvas
                ctx.clearRect(0, 0, canvas.width, canvas.height);
            }
            
            // Enable validate button only if doc uploaded
            validateBtn.disabled = !docFilename;
        });
        
        // Document upload
        docImage.addEventListener('change', async function(e) {
            const file = e.target.files[0];
            if (!file) return;
            
            // Show preview
            const reader = new FileReader();
            reader.onload = function(e) {
                documentImg.src = e.target.result;
                docPreview.style.display = 'block';
                
                // Setup canvas when image loads
                documentImg.onload = function() {
                    canvas.width = documentImg.offsetWidth;
                    canvas.height = documentImg.offsetHeight;
                    
                    // Calculate scales based on natural vs display size
                    scaleX = documentImg.naturalWidth / documentImg.offsetWidth;
                    scaleY = documentImg.naturalHeight / documentImg.offsetHeight;
                };
            };
            reader.readAsDataURL(file);
            
            // Upload to server
            const formData = new FormData();
            formData.append('file', file);
            
            try {
                const response = await fetch('/upload', { method: 'POST', body: formData });
                const data = await response.json();
                
                if (data.success) {
                    docFilename = data.filename;
                    validateBtn.disabled = false;
                    showToast('Document uploaded successfully');
                }
            } catch (error) {
                showToast('Upload failed');
            }
        });
        
        // Selfie upload
        selfieImage.addEventListener('change', async function(e) {
            const file = e.target.files[0];
            if (!file) return;
            
            // Show preview
            const reader = new FileReader();
            reader.onload = function(e) {
                selfiePreview.src = e.target.result;
                selfiePreview.style.display = 'block';
            };
            reader.readAsDataURL(file);
            
            // Upload to server
            const formData = new FormData();
            formData.append('file', file);
            
            try {
                const response = await fetch('/upload', { method: 'POST', body: formData });
                const data = await response.json();
                
                if (data.success) {
                    selfieFilename = data.filename;
                }
            } catch (error) {
                showToast('Selfie upload failed');
            }
        });
        
        // ============ REGION SELECTION WITH AUTO OCR ============
        
        canvas.addEventListener('mousedown', function(e) {
            if (validationType.value !== 'region_selection') return;
            
            isSelecting = true;
            const rect = canvas.getBoundingClientRect();
            startX = e.clientX - rect.left;
            startY = e.clientY - rect.top;
            
            // Clear any previous selection
            ctx.clearRect(0, 0, canvas.width, canvas.height);
        });
        
        canvas.addEventListener('mousemove', function(e) {
            if (!isSelecting) return;
            
            const rect = canvas.getBoundingClientRect();
            const currentX = e.clientX - rect.left;
            const currentY = e.clientY - rect.top;
            
            // Clear and draw new rectangle
            ctx.clearRect(0, 0, canvas.width, canvas.height);
            ctx.strokeStyle = '#ff0000';
            ctx.lineWidth = 2;
            ctx.setLineDash([5, 5]);
            ctx.strokeRect(startX, startY, currentX - startX, currentY - startY);
            ctx.setLineDash([]);
        });
        
        canvas.addEventListener('mouseup', async function(e) {
            if (!isSelecting || validationType.value !== 'region_selection') return;
            isSelecting = false;
            
            const rect = canvas.getBoundingClientRect();
            const endX = e.clientX - rect.left;
            const endY = e.clientY - rect.top;
            
            // Calculate region in image coordinates using scale factors
            const x = Math.min(startX, endX);
            const y = Math.min(startY, endY);
            const width = Math.abs(endX - startX);
            const height = Math.abs(endY - startY);
            
            if (width > 10 && height > 10) {
                selectedRegion = {
                    x: Math.round(x * scaleX),
                    y: Math.round(y * scaleY),
                    width: Math.round(width * scaleX),
                    height: Math.round(height * scaleY)
                };
                
                regionCoords.innerHTML = `(${selectedRegion.x}, ${selectedRegion.y}) - ${selectedRegion.width} x ${selectedRegion.height}`;
                selectedRegionDisplay.style.display = 'block';
                
                // AUTO OCR EXTRACTION - HAPPENS IMMEDIATELY ON MOUSE RELEASE
                await extractRegionAndCopyToClipboard();
            }
        });
        
        // ============ AUTO OCR EXTRACTION AND CLIPBOARD ============
        
        // ============ AUTO OCR EXTRACTION AND CLIPBOARD ============

        async function extractRegionAndCopyToClipboard() {
            if (!docFilename || !selectedRegion.width || !selectedRegion.height) {
                showToast('Please select a region first');
                return;
            }
            
            // Show loading in region info
            regionInfo.innerHTML = '⏳ Extracting text from region...';
            
            // Prepare form data
            const formData = new FormData();
            formData.append('validation_type', 'region_selection');
            formData.append('image1', docImage.files[0]);
            formData.append('region_x', selectedRegion.x);
            formData.append('region_y', selectedRegion.y);
            formData.append('region_width', selectedRegion.width);
            formData.append('region_height', selectedRegion.height);
            
            try {
                const response = await fetch('/validate', { 
                    method: 'POST', 
                    body: formData 
                });
                
                const data = await response.json();
                
                if (data.success && data.text) {
                    // Display in result box
                    resultBox.innerHTML = JSON.stringify(data, null, 2);
                    
                    // Copy to clipboard automatically
                    await copyTextToClipboard(data.text);
                    
                    // Show success message
                    showToast(`✓ Copied: "${data.text.substring(0, 50)}${data.text.length > 50 ? '...' : ''}"`);
                    
                    // Update region info
                    regionInfo.innerHTML = `✅ Region extracted: ${data.word_count} words found - Copied to clipboard!`;
                    
                    // Calculate display coordinates for highlighting
                    const displayX = selectedRegion.x / scaleX;
                    const displayY = selectedRegion.y / scaleY;
                    const displayWidth = selectedRegion.width / scaleX;
                    const displayHeight = selectedRegion.height / scaleY;
                    
                    // Highlight the region in green to show success
                    ctx.strokeStyle = '#48bb78';
                    ctx.lineWidth = 3;
                    ctx.setLineDash([]);
                    ctx.strokeRect(displayX, displayY, displayWidth, displayHeight);
                } else {
                    regionInfo.innerHTML = '❌ No text found in selected region';
                    showToast('No text found in selected region');
                }
                
            } catch (error) {
                regionInfo.innerHTML = '❌ Extraction failed';
                resultBox.innerHTML = 'Error: ' + error.message;
                showToast('Extraction failed');
            }
        }
                
        // Helper function to copy text to clipboard
        async function copyTextToClipboard(text) {
            try {
                await navigator.clipboard.writeText(text);
                return true;
            } catch (err) {
                // Fallback method
                const textarea = document.createElement('textarea');
                textarea.value = text;
                document.body.appendChild(textarea);
                textarea.select();
                document.execCommand('copy');
                document.body.removeChild(textarea);
                return true;
            }
        }
        
        // ============ VALIDATION ============
        
        validateBtn.addEventListener('click', async function() {
            const type = validationType.value;
            
            if (!type) {
                showToast('Please select validation type');
                return;
            }
            
            if (!docFilename) {
                showToast('Please upload document');
                return;
            }
            
            if (type === 'face_similarity' && !selfieFilename) {
                showToast('Please upload selfie');
                return;
            }
            
            if (type === 'keyword_match') {
                const keywords = document.getElementById('keywords').value;
                if (!keywords.trim()) {
                    showToast('Please enter keywords');
                    return;
                }
            }
            
            // Show loading
            resultBox.innerHTML = 'Processing... ⏳';
            validateBtn.disabled = true;
            
            // Prepare form data
            const formData = new FormData();
            formData.append('validation_type', type);
            formData.append('image1', docImage.files[0]);
            
            if (type === 'face_similarity' && selfieImage.files[0]) {
                formData.append('image2', selfieImage.files[0]);
            }
            
            if (type === 'keyword_match') {
                const keywords = document.getElementById('keywords').value;
                formData.append('keywords', keywords);
            }
            
            try {
                const response = await fetch('/validate', { 
                    method: 'POST', 
                    body: formData 
                });
                
                const data = await response.json();
                
                // Format result nicely
                resultBox.innerHTML = JSON.stringify(data, null, 2);
                
            } catch (error) {
                resultBox.innerHTML = 'Error: ' + error.message;
            } finally {
                validateBtn.disabled = false;
            }
        });
        
        // ============ UTILITIES ============
        
        function showToast(message) {
            toast.textContent = message;
            toast.classList.add('show');
            setTimeout(() => toast.classList.remove('show'), 2000);
        }
        
        // Resize canvas when window resizes
        window.addEventListener('resize', function() {
            if (documentImg.complete && documentImg.offsetWidth > 0) {
                canvas.width = documentImg.offsetWidth;
                canvas.height = documentImg.offsetHeight;
                
                // Recalculate scales
                scaleX = documentImg.naturalWidth / documentImg.offsetWidth;
                scaleY = documentImg.naturalHeight / documentImg.offsetHeight;
            }
        });
    </script>
</body>
</html>"""
````

````python
# KYC 16-03

from fastapi import FastAPI, File, UploadFile, Form
from fastapi.responses import HTMLResponse
from typing import List, Optional
from utils import ocr_extract, keyword_match, doc_expiry, quality_check, similarity_search
import numpy as np
import cv2
import json

app = FastAPI()

@app.get("/", response_class=HTMLResponse)
def home():
    # Fix: Specify UTF-8 encoding when reading the file
    with open("templates/index.html", "r", encoding="utf-8") as f:
        return f.read()

# This endpoint is only called for non-face-similarity validations
@app.post("/extract_text")
async def extract_text(image: UploadFile = File(...)):
    try:
        image_bytes = await image.read()
        np_image = np.frombuffer(image_bytes, np.uint8)
        img = cv2.imdecode(np_image, cv2.IMREAD_COLOR)
        
        if img is None:
            return {"error": "Failed to decode image"}
            
        raw_text = ocr_extract(img)
        image_shape = list(img.shape)
        
        return {
            "raw_text": raw_text,
            'image_shape': image_shape,
            'success': True
        }
    except Exception as e:
        return {"error": str(e), 'success': False}

@app.post("/submit")
async def submit(
    validation: str = Form(...),
    image1: UploadFile = File(...),
    image2: Optional[UploadFile] = File(None),
    keywords: Optional[List[str]] = Form(None)
):
    """
    Main submission endpoint that handles all validation types
    """
    try:
        # For face similarity, we need to pass the UploadFile objects directly
        if validation == "face_similarity":
            if image2 is None:
                return {"error": "Second image required for face similarity"}
            
            # Pass the UploadFile objects directly to similarity_search
            # Don't decode them - let similarity_search handle the file reading
            result = await similarity_search(image1, image2)
            return result
        
        # For other validations, we need to decode the image and run OCR
        else:
            # Read and decode the first image
            image1_bytes = await image1.read()
            np_image1 = np.frombuffer(image1_bytes, np.uint8)
            img1 = cv2.imdecode(np_image1, cv2.IMREAD_COLOR)
            
            if img1 is None:
                return {"error": "Failed to decode image"}
            
            # Extract text from the image
            raw_text = ocr_extract(img1)
            image_shape = list(img1.shape)
            
            # Decide which validation to run based on type
            if validation == "fraud_detection":
                if not keywords:
                    return {"error": "Keywords required for fraud detection"}
                result = keyword_match(raw_text, keywords)
                
            elif validation == "expiration_validation":
                result = doc_expiry(raw_text)
                
            elif validation == "resolution_check":
                result = quality_check(image_shape)
                
            else:
                result = {"error": f"Invalid validation type: {validation}"}
            
            return result
            
    except Exception as e:
        return {"error": f"Server error: {str(e)}"}
    
##################
# utils.py
##################

# from document_anchors import DOCUMENT_ANCHORS
import re
from rapidfuzz import process, fuzz
from rapidocr_onnxruntime import RapidOCR
import cv2, os, tempfile
from datetime import datetime
from deepface import DeepFace

ocr = RapidOCR()

def fuzzy(fields,ocr_text,threshold=70):
    corrected_list = ocr_text.copy()
    for field in fields:
        match = process.extractOne(field,corrected_list,scorer=fuzz.partial_ratio)
        if match:
            matched_word, score, index = match
            print(match)
            if score >= threshold:
                corrected_list[index] = field
    return corrected_list

def normalize(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9 ]', '', text)   # remove special characters
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def ocr_extract(image_path):
    result, _ = ocr(image_path)
    return [line[1] for line in result]

def keyword_match(raw_text,fields):
    norm_raw = [normalize(i) for i in raw_text]
    fuzzy_words = fuzzy(fields,norm_raw)
    score = 0
    for field in fields:
        if field in fuzzy_words:
            print(field)
            score += 1
    perc = (score/len(fields))*100
    return {'Score':score,'percentage':perc,'Matched':perc >= 65}

def doc_expiry(raw_text):
    dates = []
    for text in raw_text:
        matches = re.findall(r'\b\d{2}[-/]?\d{2}[-/]?\d{4}\b', text)
        for m in matches:
            clean = re.sub(r'[-/]', '', m)
            try:
                dt = datetime.strptime(clean,"%d%m%Y")
                dates.append(dt)
            except:
                pass

    max_date, today = max(dates), datetime.today()
    return 'Valid' if max_date > today else 'Expired'

def quality_check(dim):
    height, width, _ = dim[0], dim[1], dim[2]
    print(dim)
    if int(height) < 640 or int(width) < 480:
        return 'Low Resolution Image'
    else:
        return 'Good Quality Image'
    
async def similarity_search(image1, image2):
    try:
        # Reset file pointers to beginning
        await image1.seek(0)
        await image2.seek(0)
        
        # Read image bytes
        img1_bytes = await image1.read()
        img2_bytes = await image2.read()

        # Create temporary files
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp1:
            tmp1.write(img1_bytes)
            tmp1_path = tmp1.name
        
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp2:
            tmp2.write(img2_bytes)
            tmp2_path = tmp2.name
        
        try:
            # Perform face verification
            response = DeepFace.verify(
                img1_path=tmp1_path,
                img2_path=tmp2_path,
                model_name="ArcFace",
                detector_backend="retinaface",
                enforce_detection=False,
                align=True
            )
            
            # Return structured response instead of just a string
            return {
                "success": True,
                "verified": response.get("verified", False),
                "confidence": response.get("confidence", 0),
                "distance": response.get("distance", 0),
                "time": response.get("time", 0),
                "model": "ArcFace",
                "message": "Face similarity check completed"
            }
            
        except Exception as e:
            return {
                "success": False, 
                "error": f"DeepFace verification failed: {str(e)}"
            }
        finally:
            # Clean up temporary files
            try:
                os.unlink(tmp1_path)
                os.unlink(tmp2_path)
            except:
                pass
            
    except Exception as e:
        return {
            "success": False,
            "error": f"Face similarity failed: {str(e)}"
        }
    
index_html = """
<!DOCTYPE html>
<html>
<head>
    <title>Document Validation System</title>
    <style>
        * {
            box-sizing: border-box;
            font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, Oxygen, Ubuntu, Cantarell, sans-serif;
        }
        
        body {
            margin: 0;
            padding: 20px;
            min-height: 100vh;
            background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
        }

        .container {
            max-width: 700px;
            margin: 0 auto;
            background: white;
            border-radius: 20px;
            box-shadow: 0 20px 60px rgba(0,0,0,0.3);
            overflow: hidden;
        }

        .header {
            background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
            color: white;
            padding: 30px;
            text-align: center;
        }

        .header h1 {
            margin: 0;
            font-size: 28px;
            font-weight: 600;
        }

        .header p {
            margin: 10px 0 0;
            opacity: 0.9;
            font-size: 16px;
        }

        .content {
            padding: 30px;
        }

        .section {
            margin-bottom: 30px;
            background: #f8f9fa;
            border-radius: 12px;
            padding: 20px;
            border: 1px solid #e9ecef;
        }

        .section-title {
            margin: 0 0 15px 0;
            color: #495057;
            font-size: 18px;
            font-weight: 600;
            display: flex;
            align-items: center;
        }

        .section-title .step {
            background: #667eea;
            color: white;
            width: 24px;
            height: 24px;
            border-radius: 50%;
            display: inline-flex;
            align-items: center;
            justify-content: center;
            font-size: 14px;
            margin-right: 10px;
        }

        select, input[type="file"], button {
            width: 100%;
            padding: 12px 16px;
            font-size: 15px;
            border: 2px solid #e9ecef;
            border-radius: 8px;
            background: white;
            transition: all 0.3s ease;
        }

        select:focus, input[type="file"]:focus {
            outline: none;
            border-color: #667eea;
        }

        .upload-area {
            display: flex;
            flex-direction: column;
            gap: 20px;
        }

        .upload-item {
            background: white;
            border-radius: 8px;
            padding: 15px;
            border: 1px solid #e9ecef;
        }

        .upload-label {
            display: block;
            margin-bottom: 10px;
            color: #495057;
            font-weight: 500;
            font-size: 14px;
        }

        .preview-image {
            max-width: 100%;
            max-height: 200px;
            margin-top: 15px;
            border-radius: 8px;
            border: 2px solid #e9ecef;
            display: none;
            object-fit: contain;
            background: #f8f9fa;
        }

        .keyword-table {
            width: 100%;
            border-collapse: collapse;
        }

        .keyword-table td {
            padding: 5px 0;
        }

        .keyword-input {
            width: 100%;
            padding: 10px 12px;
            border: 2px solid #e9ecef;
            border-radius: 6px;
            font-size: 14px;
        }

        .keyword-input:focus {
            outline: none;
            border-color: #667eea;
        }

        .submit-btn {
            background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
            color: white;
            border: none;
            font-weight: 600;
            font-size: 16px;
            padding: 15px;
            cursor: pointer;
            margin-top: 20px;
        }

        .submit-btn:hover {
            transform: translateY(-2px);
            box-shadow: 0 10px 30px rgba(102, 126, 234, 0.4);
        }

        .submit-btn:disabled {
            opacity: 0.5;
            cursor: not-allowed;
            transform: none;
        }

        .result-section {
            background: #f8f9fa;
            border-radius: 12px;
            padding: 20px;
        }

        .result-box {
            background: white;
            border: 2px solid #e9ecef;
            border-radius: 8px;
            padding: 15px;
            min-height: 100px;
            font-family: 'Monaco', 'Menlo', monospace;
            font-size: 14px;
            white-space: pre-wrap;
            word-wrap: break-word;
            margin: 0;
        }

        .loading {
            display: inline-block;
            width: 20px;
            height: 20px;
            border: 3px solid #f3f3f3;
            border-top: 3px solid #667eea;
            border-radius: 50%;
            animation: spin 1s linear infinite;
            margin-right: 10px;
        }

        @keyframes spin {
            0% { transform: rotate(0deg); }
            100% { transform: rotate(360deg); }
        }

        .info-text {
            color: #6c757d;
            font-size: 13px;
            margin-top: 5px;
            font-style: italic;
        }

        .badge {
            background: #e9ecef;
            color: #495057;
            padding: 4px 8px;
            border-radius: 4px;
            font-size: 12px;
            font-weight: 500;
            margin-left: 10px;
        }
    </style>
</head>

<body>
    <div class="container">
        <div class="header">
            <h1>📄 Document Validation System</h1>
            <p>Upload and validate your documents with advanced AI</p>
        </div>

        <div class="content">
            <!-- Step 1: Select Validation Type -->
            <div class="section">
                <div class="section-title">
                    <span class="step">1</span>
                    Select Validation Type
                </div>
                <select id="validation-type">
                    <option value="">-- Choose a validation method --</option>
                    <option value="fraud_detection">🔍 Keyword Match</option>
                    <option value="expiration_validation">📅 Expiration Validation</option>
                    <option value="resolution_check">📏 Resolution Check</option>
                    <option value="face_similarity">👤 Face Similarity</option>
                </select>
                <div class="info-text" id="validation-info">
                    Select the type of validation you want to perform
                </div>
            </div>

            <!-- Step 2: Upload Section (Dynamic) -->
            <div class="section" id="upload-section" style="display: none;">
                <div class="section-title">
                    <span class="step">2</span>
                    Upload Images
                    <span class="badge" id="upload-badge"></span>
                </div>
                
                <!-- Single Upload (Default - for Keyword Match, Expiration, Resolution) -->
                <div id="single-upload" style="display: none;">
                    <div class="upload-area">
                        <div class="upload-item">
                            <label class="upload-label">📷 Document Image</label>
                            <input type="file" id="image1" accept="image/*" onchange="previewImage(this, 'preview1')">
                            <img class="preview-image" id="preview1" alt="Document preview">
                            <div class="info-text">Upload a clear image of your document</div>
                        </div>
                    </div>
                </div>
                
                <!-- Double Upload (For Face Similarity) -->
                <div id="double-upload" style="display: none;">
                    <div class="upload-area">
                        <div class="upload-item">
                            <label class="upload-label">📷 Document Image</label>
                            <input type="file" id="face-doc-image" accept="image/*" onchange="previewImage(this, 'preview-doc')">
                            <img class="preview-image" id="preview-doc" alt="Document preview">
                        </div>
                        
                        <div class="upload-item">
                            <label class="upload-label">🤳 Selfie Image</label>
                            <input type="file" id="face-selfie-image" accept="image/*" onchange="previewImage(this, 'preview-selfie')">
                            <img class="preview-image" id="preview-selfie" alt="Selfie preview">
                            <div class="info-text">Upload a clear selfie for face comparison</div>
                        </div>
                    </div>
                </div>
            </div>

            <!-- Step 3: Keywords Section (Only for Keyword Match) -->
            <div class="section" id="keyword-section" style="display: none;">
                <div class="section-title">
                    <span class="step">3</span>
                    Enter Keywords to Match
                </div>
                <table class="keyword-table" id="keyword-table">
                    <tr>
                        <td>
                            <input type="text" class="keyword-input" 
                                   placeholder="Type a keyword and press Enter to add more" 
                                   onkeydown="addKeywordRow(event)">
                        </td>
                    </tr>
                </table>
                <div class="info-text">Press Enter after each keyword to add another</div>
            </div>

            <!-- Submit Button -->
            <button class="submit-btn" id="submit-btn" onclick="submitData()" style="display: none;">
                🚀 Run Validation
            </button>

            <!-- Results Section -->
            <div class="section result-section">
                <div class="section-title">
                    <span class="step">📊</span>
                    Validation Result
                </div>
                <pre class="result-box" id="result">Waiting for validation...</pre>
            </div>
        </div>
    </div>

    <script>
        // Handle validation type change
        document.getElementById("validation-type").addEventListener("change", function() {
            const validationType = this.value;
            const validationInfo = document.getElementById("validation-info");
            const uploadSection = document.getElementById("upload-section");
            const submitBtn = document.getElementById("submit-btn");
            const uploadBadge = document.getElementById("upload-badge");
            
            // Reset all sections
            document.getElementById("single-upload").style.display = "none";
            document.getElementById("double-upload").style.display = "none";
            document.getElementById("keyword-section").style.display = "none";
            
            if (validationType) {
                uploadSection.style.display = "block";
                submitBtn.style.display = "block";
                
                // Update badge and info based on selection
                switch(validationType) {
                    case "fraud_detection":
                        uploadBadge.textContent = "1 image required";
                        validationInfo.textContent = "Upload a document and enter keywords to search for";
                        document.getElementById("single-upload").style.display = "block";
                        document.getElementById("keyword-section").style.display = "block";
                        break;
                        
                    case "expiration_validation":
                        uploadBadge.textContent = "1 image required";
                        validationInfo.textContent = "Upload a document to check its expiration date";
                        document.getElementById("single-upload").style.display = "block";
                        break;
                        
                    case "resolution_check":
                        uploadBadge.textContent = "1 image required";
                        validationInfo.textContent = "Upload a document to check image quality and resolution";
                        document.getElementById("single-upload").style.display = "block";
                        break;
                        
                    case "face_similarity":
                        uploadBadge.textContent = "2 images required";
                        validationInfo.textContent = "Upload both document and selfie for face comparison";
                        document.getElementById("double-upload").style.display = "block";
                        break;
                }
            } else {
                uploadSection.style.display = "none";
                submitBtn.style.display = "none";
                validationInfo.textContent = "Select the type of validation you want to perform";
            }
        });

        // Preview image function
        function previewImage(input, previewId) {
            const preview = document.getElementById(previewId);
            const file = input.files[0];
            
            if (file) {
                // Validate file type
                if (!file.type.startsWith('image/')) {
                    alert('Please upload an image file');
                    input.value = '';
                    return;
                }
                
                // Validate file size (max 10MB)
                if (file.size > 10 * 1024 * 1024) {
                    alert('File size should be less than 10MB');
                    input.value = '';
                    return;
                }
                
                const reader = new FileReader();
                reader.onload = function(e) {
                    preview.src = e.target.result;
                    preview.style.display = "block";
                };
                reader.readAsDataURL(file);
            } else {
                preview.src = "";
                preview.style.display = "none";
            }
        }

        // Add new keyword row on Enter
        function addKeywordRow(event) {
            if (event.key === "Enter") {
                event.preventDefault();
                
                const currentInput = event.target;
                if (currentInput.value.trim() === "") {
                    return; // Don't add empty keywords
                }
                
                const table = document.getElementById("keyword-table");
                const newRow = table.insertRow(-1);
                const newCell = newRow.insertCell(0);
                
                newCell.innerHTML = '<input type="text" class="keyword-input" placeholder="Type another keyword" onkeydown="addKeywordRow(event)">';
                newCell.firstChild.focus();
            }
        }

        // Get all keywords from input fields
        function getKeywords() {
            const keywords = [];
            document.querySelectorAll(".keyword-input").forEach(input => {
                if (input.value.trim() !== "") {
                    keywords.push(input.value.trim());
                }
            });
            return keywords;
        }

        // Submit data
        async function submitData() {
            const validationType = document.getElementById("validation-type").value;
            const submitBtn = document.getElementById("submit-btn");
            
            if (!validationType) {
                alert("Please select a validation type");
                return;
            }

            const formData = new FormData();
            formData.append("validation", validationType);

            // Handle file uploads based on validation type
            if (validationType === "face_similarity") {
                // Face similarity - need both images
                const docImage = document.getElementById("face-doc-image").files[0];
                const selfieImage = document.getElementById("face-selfie-image").files[0];
                
                if (!docImage || !selfieImage) {
                    alert("Please upload both document and selfie images");
                    return;
                }
                
                formData.append("image1", docImage);
                formData.append("image2", selfieImage);
                
            } else {
                // Other validations - only one image needed
                const image1 = document.getElementById("image1").files[0];
                
                if (!image1) {
                    alert("Please upload document image");
                    return;
                }
                
                formData.append("image1", image1);
                
                // Add keywords for fraud detection
                if (validationType === "fraud_detection") {
                    const keywords = getKeywords();
                    
                    if (keywords.length === 0) {
                        alert("Please enter at least one keyword");
                        return;
                    }
                    
                    keywords.forEach(keyword => {
                        formData.append("keywords", keyword);
                    });
                }
            }

            // Show loading state
            const resultElement = document.getElementById("result");
            const originalText = resultElement.innerText;
            resultElement.innerHTML = '<span class="loading"></span> Processing your request...';
            submitBtn.disabled = true;
            submitBtn.style.opacity = "0.5";
            
            try {
                // Send request
                const response = await fetch("/submit", {
                    method: "POST",
                    body: formData
                });
                
                const data = await response.json();
                resultElement.innerHTML = JSON.stringify(data, null, 2);
                
            } catch (error) {
                resultElement.innerHTML = '❌ Error: ' + error.message;
            } finally {
                submitBtn.disabled = false;
                submitBtn.style.opacity = "1";
            }
        }

        // Clear all inputs when page loads
        window.onload = function() {
            document.getElementById("validation-type").value = "";
            
            // Add enter key handler for the first input
            const firstInput = document.querySelector('.keyword-input');
            if (firstInput) {
                firstInput.addEventListener('keydown', addKeywordRow);
            }
        };
    </script>
</body>
</html>
"""
````

````python
# KYC 20-03

from fastapi import FastAPI, File, UploadFile, Form, Request
from fastapi.responses import HTMLResponse, JSONResponse, FileResponse
from fastapi.templating import Jinja2Templates
import shutil, os, uuid, cv2, numpy as np, tempfile, re, uvicorn
from pathlib import Path
from datetime import datetime
from rapidfuzz import process, fuzz
from rapidocr_onnxruntime import RapidOCR
from deepface import DeepFace

app = FastAPI()
UPLOAD_DIR = Path("uploads")
UPLOAD_DIR.mkdir(exist_ok=True)
templates = Jinja2Templates(directory="templates")

# Initialize OCR
ocr = RapidOCR()

# ============ HELPER FUNCTIONS ============

def extract_text(image_path):
    """Extract text from image"""
    result, _ = ocr(image_path)
    return [line[1] for line in result] if result else []

def extract_text_from_cropped(image_array):
    """Extract text from cropped image array"""
    # Save cropped image temporarily
    with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp:
        cv2.imwrite(tmp.name, image_array)
        tmp_path = tmp.name
    
    try:
        result, _ = ocr(tmp_path)
        return [line[1] for line in result] if result else []
    finally:
        os.unlink(tmp_path)

def fuzzy_match(text_list, keywords, threshold=70):
    """Fuzzy match keywords"""
    corrected = text_list.copy()
    for keyword in keywords:
        match = process.extractOne(keyword, corrected, scorer=fuzz.partial_ratio)
        if match and match[1] >= threshold:
            corrected[match[2]] = keyword
    return corrected

def keyword_match(text_list, keywords):
    """Check if keywords exist in text"""
    normalized = [t.lower() for t in text_list]
    matched = [k for k in keywords if any(k.lower() in t for t in normalized)]
    return {
        "found": matched,
        "not_found": [k for k in keywords if k not in matched],
        "score": len(matched),
        "percentage": (len(matched) / len(keywords)) * 100 if keywords else 0,
        "matched": (len(matched) / len(keywords)) * 100 >= 65 if keywords else False
    }

def find_expiry(text_list):
    """Find expiration date"""
    dates = []
    for text in text_list:
        matches = re.findall(r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', text)
        for m in matches:
            try:
                clean = re.sub(r'[-/]', '', m)
                dates.append(datetime.strptime(clean, "%d%m%Y"))
            except:
                pass
    if not dates:
        return "No expiry date found"
    max_date = max(dates)
    return "Valid" if max_date > datetime.today() else "Expired"

def check_resolution(img_path):
    """Check image quality"""
    img = cv2.imread(img_path)
    h, w = img.shape[:2]
    if h < 640 or w < 480:
        return f"Low Resolution ({w}x{h})"
    return f"Good Quality ({w}x{h})"

async def compare_faces(img1, img2):
    """Compare two faces"""
    # Save temp files
    with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as t1:
        t1.write(img1)
        t1_path = t1.name
    with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as t2:
        t2.write(img2)
        t2_path = t2.name
    
    try:
        result = DeepFace.verify(
            img1_path=t1_path,
            img2_path=t2_path,
            model_name="ArcFace",
            enforce_detection=False
        )
        return {
            "match": result.get("verified", False),
            "confidence": result.get("confidence", 0),
            "distance": result.get("distance", 0)
        }
    finally:
        os.unlink(t1_path)
        os.unlink(t2_path)

# ============ API ENDPOINTS ============

@app.get("/", response_class=HTMLResponse)
async def home(request: Request):
    return templates.TemplateResponse("index.html", {"request": request})

@app.post("/upload")
async def upload(file: UploadFile = File(...)):
    filename = f"{uuid.uuid4()}{Path(file.filename).suffix}"
    path = UPLOAD_DIR / filename
    with open(path, "wb") as f:
        shutil.copyfileobj(file.file, f)
    return {"success": True, "filename": filename}

@app.post("/process")
async def process(
    type: str = Form(...),
    image1: UploadFile = File(...),
    image2: UploadFile = File(None),
    keywords: str = Form(""),
    region_x: int = Form(0),
    region_y: int = Form(0),
    region_w: int = Form(0),
    region_h: int = Form(0)
):
    try:
        # Read first image bytes
        img1_bytes = await image1.read()
        
        # Handle face similarity
        if type == "face_similarity":
            if not image2:
                return {"error": "Second image required"}
            img2_bytes = await image2.read()
            result = await compare_faces(img1_bytes, img2_bytes)
            return {"type": "face_similarity", **result}
        
        # For region selection, we need to crop and process
        if type == "region_selection":
            if region_w > 0 and region_h > 0:
                # Convert bytes to image
                nparr = np.frombuffer(img1_bytes, np.uint8)
                img = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
                
                if img is None:
                    return {"error": "Could not decode image"}
                
                # Check if region is within bounds
                h, w = img.shape[:2]
                if region_x + region_w > w or region_y + region_h > h:
                    return {"error": f"Region out of bounds. Image size: {w}x{h}, Region: {region_x},{region_y} {region_w}x{region_h}"}
                
                # Crop the region
                cropped = img[region_y:region_y+region_h, region_x:region_x+region_w]
                
                # Extract text from cropped region
                texts = extract_text_from_cropped(cropped)
                full_text = " ".join(texts)
                
                return {
                    "type": "region_selection",
                    "text": full_text,
                    "words": len(texts),
                    "region": {"x": region_x, "y": region_y, "width": region_w, "height": region_h}
                }
            return {"error": "Invalid region dimensions"}
        
        # For other types, save image temporarily
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp:
            tmp.write(img1_bytes)
            img_path = tmp.name
        
        try:
            # Extract all text
            all_text = extract_text(img_path)
            
            # Keyword match
            if type == "keyword_match":
                kw_list = [k.strip() for k in keywords.split(",") if k.strip()]
                if not kw_list:
                    return {"error": "No keywords provided"}
                return {
                    "type": "keyword_match",
                    **keyword_match(all_text, kw_list),
                    "all_text": " ".join(all_text)[:500]
                }
            
            # Expiration
            if type == "expiration":
                return {
                    "type": "expiration",
                    "status": find_expiry(all_text),
                    "all_text": " ".join(all_text)[:500]
                }
            
            # Resolution
            if type == "resolution":
                return {
                    "type": "resolution",
                    "quality": check_resolution(img_path)
                }
            
            return {"error": "Invalid type"}
            
        finally:
            os.unlink(img_path)
            
    except Exception as e:
        return {"error": str(e)}

@app.get("/image/{filename}")
async def get_image(filename: str):
    return FileResponse(UPLOAD_DIR / filename)

index_html='''<!DOCTYPE html>
<html>
<head>
    <title>Document Validation</title>
    <style>
        * { margin: 0; padding: 0; box-sizing: border-box; font-family: Arial, sans-serif; }
        body { background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); min-height: 100vh; padding: 20px; }
        .container { max-width: 1200px; margin: 0 auto; background: white; border-radius: 15px; padding: 30px; }
        h1 { margin-bottom: 10px; color: #333; }
        .card { background: #f8f9fa; padding: 20px; border-radius: 10px; margin-bottom: 20px; }
        select, input, button { width: 100%; padding: 12px; margin: 10px 0; border: 2px solid #ddd; border-radius: 8px; font-size: 14px; }
        button { background: #48bb78; color: white; border: none; cursor: pointer; font-weight: bold; }
        button:hover { background: #38a169; }
        button:disabled { background: #ccc; cursor: not-allowed; }
        .image-container { position: relative; margin: 20px 0; display: none; background: #f0f0f0; border-radius: 8px; overflow: hidden; }
        #documentImage { width: 100%; height: auto; display: block; }
        #selectionCanvas { position: absolute; top: 0; left: 0; width: 100%; height: 100%; cursor: crosshair; pointer-events: none; }
        #selectionCanvas.active { pointer-events: all; }
        .result-box { background: #f0f0f0; padding: 15px; border-radius: 8px; font-family: monospace; white-space: pre-wrap; max-height: 400px; overflow-y: auto; margin-top: 20px; font-size: 12px; }
        .toast { position: fixed; bottom: 30px; right: 30px; background: #333; color: white; padding: 12px 24px; border-radius: 8px; display: none; z-index: 1000; }
        .toast.show { display: block; }
        .info-text { color: #666; font-size: 12px; margin-top: 5px; }
        #selfiePreview { max-width: 100%; margin-top: 10px; display: none; border-radius: 8px; }
        .region-info { padding: 10px; background: #e3f2fd; border-radius: 8px; margin: 10px 0; font-size: 14px; }
    </style>
</head>
<body>
    <div class="container">
        <h1>📄 Document Validation System</h1>
        
        <!-- Validation Type -->
        <div class="card">
            <select id="type">
                <option value="">Select Validation Type</option>
                <option value="keyword_match">Keyword Match</option>
                <option value="expiration">Expiration Date</option>
                <option value="resolution">Resolution Check</option>
                <option value="face_similarity">Face Similarity</option>
                <option value="region_selection">Select Region (Auto-copy)</option>
            </select>
        </div>
        
        <!-- Upload Section -->
        <div class="card">
            <input type="file" id="docImage" accept="image/*">
            <div class="image-container" id="docPreview">
                <img id="documentImage" src="">
                <canvas id="selectionCanvas"></canvas>
            </div>
            <div id="selfieSection" style="display: none;">
                <input type="file" id="selfieImage" accept="image/*">
                <img id="selfiePreview">
            </div>
        </div>
        
        <!-- Keywords (for keyword match) -->
        <div class="card" id="keywordSection" style="display: none;">
            <input type="text" id="keywords" placeholder="Enter keywords (comma separated)">
            <div class="info-text">Example: name, date, passport, address</div>
        </div>
        
        <!-- Region Info -->
        <div id="regionInfo" style="display: none;" class="region-info"></div>
        
        <!-- Process Button -->
        <button id="processBtn" disabled>Process</button>
        
        <!-- Results -->
        <div class="card">
            <div id="result" class="result-box">Waiting for input...</div>
        </div>
    </div>
    
    <div class="toast" id="toast"></div>
    
    <script>
        let filename = null;
        let scaleX = 1, scaleY = 1;
        let startX, startY, endX, endY;
        let isSelecting = false;
        
        const typeSelect = document.getElementById('type');
        const docImage = document.getElementById('docImage');
        const selfieImage = document.getElementById('selfieImage');
        const documentImg = document.getElementById('documentImage');
        const canvas = document.getElementById('selectionCanvas');
        const ctx = canvas.getContext('2d');
        const docPreview = document.getElementById('docPreview');
        const selfiePreview = document.getElementById('selfiePreview');
        const selfieSection = document.getElementById('selfieSection');
        const keywordSection = document.getElementById('keywordSection');
        const regionInfo = document.getElementById('regionInfo');
        const processBtn = document.getElementById('processBtn');
        const resultBox = document.getElementById('result');
        const toast = document.getElementById('toast');
        
        // Show toast message
        function showToast(msg) {
            toast.textContent = msg;
            toast.classList.add('show');
            setTimeout(() => toast.classList.remove('show'), 2000);
        }
        
        // Upload file
        async function uploadFile(file) {
            const formData = new FormData();
            formData.append('file', file);
            const res = await fetch('/upload', { method: 'POST', body: formData });
            const data = await res.json();
            if (data.success) return data.filename;
            throw new Error('Upload failed');
        }
        
        // Process validation
        async function processRequest(data) {
            const formData = new FormData();
            for (let key in data) {
                formData.append(key, data[key]);
            }
            const res = await fetch('/process', { method: 'POST', body: formData });
            return await res.json();
        }
        
        // Copy to clipboard
        async function copyToClipboard(text) {
            try {
                await navigator.clipboard.writeText(text);
                showToast(`✓ Copied: ${text.substring(0, 50)}${text.length > 50 ? '...' : ''}`);
            } catch(err) {
                // Fallback
                const textarea = document.createElement('textarea');
                textarea.value = text;
                document.body.appendChild(textarea);
                textarea.select();
                document.execCommand('copy');
                document.body.removeChild(textarea);
                showToast(`✓ Copied: ${text.substring(0, 50)}${text.length > 50 ? '...' : ''}`);
            }
        }
        
        // Handle document upload
        docImage.addEventListener('change', async (e) => {
            const file = e.target.files[0];
            if (!file) return;
            
            // Preview
            const reader = new FileReader();
            reader.onload = (e) => {
                documentImg.src = e.target.result;
                docPreview.style.display = 'block';
                documentImg.onload = () => {
                    canvas.width = documentImg.offsetWidth;
                    canvas.height = documentImg.offsetHeight;
                    scaleX = documentImg.naturalWidth / documentImg.offsetWidth;
                    scaleY = documentImg.naturalHeight / documentImg.offsetHeight;
                };
            };
            reader.readAsDataURL(file);
            
            // Upload
            try {
                filename = await uploadFile(file);
                processBtn.disabled = false;
                showToast('Document uploaded');
            } catch (err) {
                showToast('Upload failed');
            }
        });
        
        // Handle selfie upload
        selfieImage.addEventListener('change', async (e) => {
            const file = e.target.files[0];
            if (!file) return;
            const reader = new FileReader();
            reader.onload = (e) => {
                selfiePreview.src = e.target.result;
                selfiePreview.style.display = 'block';
            };
            reader.readAsDataURL(file);
            await uploadFile(file);
        });
        
        // Handle validation type change
        typeSelect.addEventListener('change', () => {
            const type = typeSelect.value;
            selfieSection.style.display = type === 'face_similarity' ? 'block' : 'none';
            keywordSection.style.display = type === 'keyword_match' ? 'block' : 'none';
            
            if (type === 'region_selection') {
                canvas.classList.add('active');
                regionInfo.style.display = 'block';
                regionInfo.innerHTML = 'Drag mouse on the image to select a region - Text will auto-copy!';
            } else {
                canvas.classList.remove('active');
                regionInfo.style.display = 'none';
                ctx.clearRect(0, 0, canvas.width, canvas.height);
            }
        });
        
        // Region selection
        canvas.addEventListener('mousedown', (e) => {
            if (typeSelect.value !== 'region_selection') return;
            isSelecting = true;
            const rect = canvas.getBoundingClientRect();
            startX = e.clientX - rect.left;
            startY = e.clientY - rect.top;
            ctx.clearRect(0, 0, canvas.width, canvas.height);
        });
        
        canvas.addEventListener('mousemove', (e) => {
            if (!isSelecting) return;
            const rect = canvas.getBoundingClientRect();
            endX = e.clientX - rect.left;
            endY = e.clientY - rect.top;
            ctx.clearRect(0, 0, canvas.width, canvas.height);
            ctx.strokeStyle = '#ff0000';
            ctx.lineWidth = 2;
            ctx.strokeRect(startX, startY, endX - startX, endY - startY);
        });
        
        canvas.addEventListener('mouseup', async (e) => {
            if (!isSelecting || typeSelect.value !== 'region_selection') return;
            isSelecting = false;
            
            const rect = canvas.getBoundingClientRect();
            endX = e.clientX - rect.left;
            endY = e.clientY - rect.top;
            
            const x = Math.min(startX, endX);
            const y = Math.min(startY, endY);
            const w = Math.abs(endX - startX);
            const h = Math.abs(endY - startY);
            
            if (w > 10 && h > 10) {
                regionInfo.innerHTML = 'Extracting text from selected region...';
                
                // Convert display coordinates to actual image coordinates
                const actualX = Math.round(x * scaleX);
                const actualY = Math.round(y * scaleY);
                const actualW = Math.round(w * scaleX);
                const actualH = Math.round(h * scaleY);
                
                const result = await processRequest({
                    type: 'region_selection',
                    image1: docImage.files[0],
                    region_x: actualX,
                    region_y: actualY,
                    region_w: actualW,
                    region_h: actualH
                });
                
                console.log('Result:', result);
                
                if (result.text && result.text.trim()) {
                    resultBox.innerHTML = JSON.stringify(result, null, 2);
                    await copyToClipboard(result.text);
                    regionInfo.innerHTML = `Extracted ${result.words} words! Copied to clipboard.`;
                    regionInfo.style.background = '#d4edda';
                    
                    // Highlight in green
                    ctx.strokeStyle = '#48bb78';
                    ctx.lineWidth = 3;
                    ctx.strokeRect(startX, startY, w, h);
                    
                    // Fill with light green
                    ctx.fillStyle = 'rgba(72, 187, 120, 0.1)';
                    ctx.fillRect(startX, startY, w, h);
                } else {
                    regionInfo.innerHTML = 'No text found in selected region. Try selecting a different area.';
                    regionInfo.style.background = '#f8d7da';
                    
                    // Highlight in red
                    ctx.strokeStyle = '#dc3545';
                    ctx.lineWidth = 2;
                    ctx.strokeRect(startX, startY, w, h);
                }
            } else {
                regionInfo.innerHTML = 'Selection too small. Please drag a larger area.';
            }
        });
        
        // Process button click
        processBtn.addEventListener('click', async () => {
            const type = typeSelect.value;
            if (!type) { showToast('Select validation type'); return; }
            if (!filename) { showToast('Upload document'); return; }
            if (type === 'face_similarity' && !selfieImage.files[0]) {
                showToast('Upload selfie');
                return;
            }
            if (type === 'keyword_match' && !document.getElementById('keywords').value.trim()) {
                showToast('Enter keywords');
                return;
            }
            
            resultBox.innerHTML = 'Processing...';
            processBtn.disabled = true;
            
            const data = {
                type: type,
                image1: docImage.files[0]
            };
            
            if (type === 'face_similarity') data.image2 = selfieImage.files[0];
            if (type === 'keyword_match') data.keywords = document.getElementById('keywords').value;
            
            const result = await processRequest(data);
            resultBox.innerHTML = JSON.stringify(result, null, 2);
            processBtn.disabled = false;
        });
        
        // Clear region when changing type
        typeSelect.addEventListener('change', () => {
            if (typeSelect.value !== 'region_selection') {
                ctx.clearRect(0, 0, canvas.width, canvas.height);
                regionInfo.style.display = 'none';
            }
        });
    </script>
</body>
</html>'''

````

````python
# PyQt5_KYC version
import sys
import cv2
import numpy as np
import tempfile
import os
import re
from datetime import datetime
from PyQt5.QtWidgets import (QApplication, QMainWindow, QWidget, QVBoxLayout, 
                             QHBoxLayout, QLabel, QPushButton, QFileDialog, 
                             QComboBox, QTextEdit, QLineEdit, QSplitter, 
                             QGroupBox, QMessageBox, QProgressBar, QScrollArea,
                             QFrame)
from PyQt5.QtCore import Qt, QThread, pyqtSignal, QRect
from PyQt5.QtGui import QPixmap, QImage, QPainter, QPen, QColor, QFont
from rapidfuzz import process, fuzz
from rapidocr_onnxruntime import RapidOCR
from deepface import DeepFace
import numpy as np

# Initialize OCR
ocr = RapidOCR()

# ============ HELPER FUNCTIONS ============

def extract_text_from_image(image_array):
    """Extract text from image using RapidOCR"""
    try:
        with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as tmp:
            cv2.imwrite(tmp.name, image_array)
            tmp_path = tmp.name
        
        try:
            result, _ = ocr(tmp_path)
            return [line[1] for line in result] if result else []
        finally:
            os.unlink(tmp_path)
    except Exception as e:
        print(f"OCR Error: {e}")
        return []

def keyword_match(text_list, keywords):
    """Check if keywords exist in text"""
    normalized = [t.lower() for t in text_list]
    matched = []
    for keyword in keywords:
        if any(keyword.lower() in t for t in normalized):
            matched.append(keyword)
    
    return {
        "found": matched,
        "not_found": [k for k in keywords if k not in matched],
        "score": len(matched),
        "total": len(keywords),
        "percentage": (len(matched) / len(keywords)) * 100 if keywords else 0,
        "matched": (len(matched) / len(keywords)) * 100 >= 65 if keywords else False
    }

def find_expiry(text_list):
    """Find expiration date"""
    dates = []
    for text in text_list:
        matches = re.findall(r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', text)
        for m in matches:
            try:
                clean = re.sub(r'[-/]', '', m)
                dates.append(datetime.strptime(clean, "%d%m%Y"))
            except:
                pass
    if not dates:
        return "No expiry date found"
    max_date = max(dates)
    return "Valid" if max_date > datetime.today() else f"Expired on {max_date.strftime('%d/%m/%Y')}"

def check_resolution(image_array):
    """Check image quality"""
    h, w = image_array.shape[:2]
    if h < 640 or w < 480:
        return f"Low Resolution ({w}x{h})"
    return f"Good Quality ({w}x{h})"

def compare_faces(img1_path, img2_path):
    """Compare two faces"""
    try:
        result = DeepFace.verify(
            img1_path=img1_path,
            img2_path=img2_path,
            model_name="ArcFace",
            enforce_detection=False
        )
        return {
            "match": result.get("verified", False),
            "confidence": result.get("confidence", 0),
            "distance": result.get("distance", 0)
        }
    except Exception as e:
        return {"error": str(e)}

# ============ WORKER THREAD ============

class OCRWorker(QThread):
    finished = pyqtSignal(dict)
    progress = pyqtSignal(str)
    
    def __init__(self, validation_type, doc_img, selfie_img=None, keywords=None, 
                 region=None):
        super().__init__()
        self.validation_type = validation_type
        self.doc_img = doc_img
        self.selfie_img = selfie_img
        self.keywords = keywords
        self.region = region
    
    def run(self):
        try:
            if self.validation_type == "keyword_match":
                self.progress.emit("Extracting text from document...")
                texts = extract_text_from_image(self.doc_img)
                self.progress.emit(f"Found {len(texts)} text segments")
                result = keyword_match(texts, self.keywords)
                result["all_text"] = "\n".join(texts[:50]) if texts else "No text found"
                self.finished.emit(result)
                
            elif self.validation_type == "expiration":
                self.progress.emit("Extracting text for expiration date...")
                texts = extract_text_from_image(self.doc_img)
                self.progress.emit(f"Found {len(texts)} text segments")
                result = find_expiry(texts)
                self.finished.emit({"status": result, "all_text": "\n".join(texts[:50]) if texts else "No text found"})
                
            elif self.validation_type == "resolution":
                self.progress.emit("Checking image resolution...")
                result = check_resolution(self.doc_img)
                h, w = self.doc_img.shape[:2]
                self.finished.emit({"quality": result, "width": w, "height": h})
                
            elif self.validation_type == "face_similarity":
                self.progress.emit("Comparing faces...")
                with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as t1:
                    cv2.imwrite(t1.name, self.doc_img)
                    t1_path = t1.name
                with tempfile.NamedTemporaryFile(delete=False, suffix=".jpg") as t2:
                    cv2.imwrite(t2.name, self.selfie_img)
                    t2_path = t2.name
                try:
                    result = compare_faces(t1_path, t2_path)
                    self.finished.emit(result)
                finally:
                    os.unlink(t1_path)
                    os.unlink(t2_path)
                    
            elif self.validation_type == "region_selection" and self.region:
                self.progress.emit("Extracting text from selected region...")
                x, y, w, h = self.region
                cropped = self.doc_img[y:y+h, x:x+w]
                texts = extract_text_from_image(cropped)
                full_text = " ".join(texts)
                self.finished.emit({"text": full_text, "words": len(texts)})
                
        except Exception as e:
            import traceback
            traceback.print_exc()
            self.finished.emit({"error": str(e)})

# ============ IMAGE VIEWER WITH REGION SELECTION ============

class ImageViewer(QLabel):
    region_selected = pyqtSignal(int, int, int, int)
    
    def __init__(self):
        super().__init__()
        self.setMinimumSize(400, 300)
        self.setStyleSheet("border: 2px solid #ddd; background-color: #f0f0f0;")
        self.setAlignment(Qt.AlignCenter)
        self.setScaledContents(False)
        self.pixmap = None
        self.original_image = None
        self.scale_factor = 1.0
        self.start_point = None
        self.end_point = None
        self.drawing = False
        self.region_mode = False
        
    def set_image(self, cv_image):
        """Set image from OpenCV array"""
        self.original_image = cv_image.copy()
        rgb_image = cv2.cvtColor(cv_image, cv2.COLOR_BGR2RGB)
        h, w, ch = rgb_image.shape
        bytes_per_line = ch * w
        qt_image = QImage(rgb_image.data, w, h, bytes_per_line, QImage.Format_RGB888)
        self.pixmap = QPixmap.fromImage(qt_image)
        self.update_display()
        
    def update_display(self):
        """Update the displayed image"""
        if self.pixmap:
            scaled_pixmap = self.pixmap.scaled(self.size(), Qt.KeepAspectRatio, 
                                                Qt.SmoothTransformation)
            self.setPixmap(scaled_pixmap)
            self.scale_factor = scaled_pixmap.width() / self.pixmap.width()
        
    def enable_region_selection(self, enabled):
        """Enable/disable region selection mode"""
        self.region_mode = enabled
        if not enabled:
            self.start_point = None
            self.end_point = None
            self.drawing = False
            self.update()
            
    def get_image_coordinates(self, pos):
        """Convert widget coordinates to image coordinates"""
        if not self.pixmap:
            return None, None
            
        # Get pixmap rectangle
        pixmap_rect = self.pixmap.rect()
        pixmap_rect.moveCenter(self.rect().center())
        
        # Calculate if point is inside pixmap
        if not pixmap_rect.contains(pos):
            return None, None
            
        # Convert to image coordinates
        x = int((pos.x() - pixmap_rect.x()) / self.scale_factor)
        y = int((pos.y() - pixmap_rect.y()) / self.scale_factor)
        
        # Clamp to image bounds
        x = max(0, min(x, self.pixmap.width() - 1))
        y = max(0, min(y, self.pixmap.height() - 1))
        
        return x, y
        
    def mousePressEvent(self, event):
        if self.region_mode and event.button() == Qt.LeftButton:
            x, y = self.get_image_coordinates(event.pos())
            if x is not None:
                self.start_point = (x, y)
                self.drawing = True
            
    def mouseMoveEvent(self, event):
        if self.drawing and self.start_point:
            x, y = self.get_image_coordinates(event.pos())
            if x is not None:
                self.end_point = (x, y)
                self.update()
            
    def mouseReleaseEvent(self, event):
        if self.drawing and self.start_point and self.end_point:
            x1, y1 = self.start_point
            x2, y2 = self.end_point
            
            x = min(x1, x2)
            y = min(y1, y2)
            w = abs(x2 - x1)
            h = abs(y2 - y1)
            
            if w > 10 and h > 10:
                self.region_selected.emit(x, y, w, h)
                
            self.drawing = False
            self.start_point = None
            self.end_point = None
            self.update()
            
    def paintEvent(self, event):
        super().paintEvent(event)
        if self.drawing and self.start_point and self.end_point:
            painter = QPainter(self)
            pen = QPen(QColor(255, 0, 0), 2, Qt.DashLine)
            painter.setPen(pen)
            
            # Convert image coordinates to widget coordinates
            if self.pixmap:
                pixmap_rect = self.pixmap.rect()
                pixmap_rect.moveCenter(self.rect().center())
                
                x1 = int(self.start_point[0] * self.scale_factor + pixmap_rect.x())
                y1 = int(self.start_point[1] * self.scale_factor + pixmap_rect.y())
                x2 = int(self.end_point[0] * self.scale_factor + pixmap_rect.x())
                y2 = int(self.end_point[1] * self.scale_factor + pixmap_rect.y())
                
                rect = QRect(x1, y1, x2 - x1, y2 - y1)
                painter.drawRect(rect)
                
    def resizeEvent(self, event):
        super().resizeEvent(event)
        self.update_display()

# ============ MAIN APPLICATION ============

class DocumentValidator(QMainWindow):
    def __init__(self):
        super().__init__()
        self.doc_img = None
        self.selfie_img = None
        self.worker = None
        self.init_ui()
        
    def init_ui(self):
        self.setWindowTitle("Document Validation System")
        self.setGeometry(100, 100, 1400, 900)
        
        # Central widget
        central_widget = QWidget()
        self.setCentralWidget(central_widget)
        main_layout = QHBoxLayout(central_widget)
        
        # Left panel - Image viewer
        left_panel = QWidget()
        left_layout = QVBoxLayout(left_panel)
        
        # Document image
        left_layout.addWidget(QLabel("📄 Document Image:"))
        self.image_viewer = ImageViewer()
        left_layout.addWidget(self.image_viewer)
        
        # Selfie image (for face similarity)
        self.selfie_label = QLabel("🤳 Selfie Image:")
        self.selfie_label.setVisible(False)
        left_layout.addWidget(self.selfie_label)
        
        self.selfie_viewer = ImageViewer()
        self.selfie_viewer.setVisible(False)
        left_layout.addWidget(self.selfie_viewer)
        
        # Upload buttons
        btn_layout = QHBoxLayout()
        self.upload_btn = QPushButton("📁 Upload Document")
        self.upload_btn.clicked.connect(self.upload_document)
        btn_layout.addWidget(self.upload_btn)
        
        self.upload_selfie_btn = QPushButton("📁 Upload Selfie")
        self.upload_selfie_btn.clicked.connect(self.upload_selfie)
        self.upload_selfie_btn.setVisible(False)
        btn_layout.addWidget(self.upload_selfie_btn)
        
        left_layout.addLayout(btn_layout)
        
        # Right panel - Controls and results
        right_panel = QScrollArea()
        right_panel.setWidgetResizable(True)
        right_widget = QWidget()
        right_layout = QVBoxLayout(right_widget)
        
        # Validation type
        right_layout.addWidget(QLabel("🔍 Validation Type:"))
        self.type_combo = QComboBox()
        self.type_combo.addItems([
            "Keyword Match",
            "Expiration Date",
            "Resolution Check",
            "Face Similarity",
            "Select Region"
        ])
        self.type_combo.currentTextChanged.connect(self.on_type_changed)
        right_layout.addWidget(self.type_combo)
        
        # Keywords input
        self.keywords_group = QGroupBox("📝 Keywords (one per line)")
        self.keywords_group.setVisible(False)
        keywords_layout = QVBoxLayout()
        self.keywords_input = QTextEdit()
        self.keywords_input.setPlaceholderText("Example:\nname\ndate\npassport\naddress")
        self.keywords_input.setMaximumHeight(150)
        keywords_layout.addWidget(self.keywords_input)
        self.keywords_group.setLayout(keywords_layout)
        right_layout.addWidget(self.keywords_group)
        
        # Region info
        self.region_info = QLabel("")
        self.region_info.setVisible(False)
        self.region_info.setWordWrap(True)
        self.region_info.setStyleSheet("background-color: #e3f2fd; padding: 10px; border-radius: 5px;")
        right_layout.addWidget(self.region_info)
        
        # Progress bar
        self.progress_bar = QProgressBar()
        self.progress_bar.setVisible(False)
        right_layout.addWidget(self.progress_bar)
        
        # Process button
        self.process_btn = QPushButton("🚀 Process Validation")
        self.process_btn.clicked.connect(self.process_validation)
        self.process_btn.setEnabled(False)
        self.process_btn.setStyleSheet("QPushButton { background-color: #4CAF50; color: white; padding: 10px; font-size: 14px; }")
        right_layout.addWidget(self.process_btn)
        
        # Results
        right_layout.addWidget(QLabel("📊 Results:"))
        self.results_text = QTextEdit()
        self.results_text.setReadOnly(True)
        self.results_text.setFont(QFont("Courier", 10))
        self.results_text.setMinimumHeight(300)
        right_layout.addWidget(self.results_text)
        
        right_panel.setWidget(right_widget)
        
        # Add to main layout
        splitter = QSplitter(Qt.Horizontal)
        splitter.addWidget(left_panel)
        splitter.addWidget(right_panel)
        splitter.setSizes([700, 500])
        main_layout.addWidget(splitter)
        
        # Connect signals
        self.image_viewer.region_selected.connect(self.on_region_selected)
        
        # Status bar
        self.statusBar().showMessage("Ready")
        
    def upload_document(self):
        file_path, _ = QFileDialog.getOpenFileName(self, "Select Document", "", 
                                                    "Images (*.png *.jpg *.jpeg *.bmp)")
        if file_path:
            try:
                self.doc_img = cv2.imread(file_path)
                if self.doc_img is None:
                    QMessageBox.critical(self, "Error", "Could not load image")
                    return
                self.image_viewer.set_image(self.doc_img)
                self.process_btn.setEnabled(True)
                self.statusBar().showMessage(f"Document loaded: {file_path}")
                QMessageBox.information(self, "Success", "Document uploaded successfully!")
            except Exception as e:
                QMessageBox.critical(self, "Error", f"Failed to load image: {str(e)}")
            
    def upload_selfie(self):
        file_path, _ = QFileDialog.getOpenFileName(self, "Select Selfie", "", 
                                                    "Images (*.png *.jpg *.jpeg *.bmp)")
        if file_path:
            try:
                self.selfie_img = cv2.imread(file_path)
                if self.selfie_img is None:
                    QMessageBox.critical(self, "Error", "Could not load image")
                    return
                self.selfie_viewer.set_image(self.selfie_img)
                self.selfie_viewer.setVisible(True)
                self.selfie_label.setVisible(True)
                self.statusBar().showMessage(f"Selfie loaded: {file_path}")
            except Exception as e:
                QMessageBox.critical(self, "Error", f"Failed to load selfie: {str(e)}")
            
    def on_type_changed(self, text):
        """Handle validation type change"""
        self.keywords_group.setVisible(text == "Keyword Match")
        self.upload_selfie_btn.setVisible(text == "Face Similarity")
        self.region_info.setVisible(text == "Select Region")
        
        # Reset region selection mode
        if text == "Select Region":
            self.image_viewer.enable_region_selection(True)
            self.region_info.setText("✂️ Click and drag on the document image to select a region. Text will be automatically extracted and copied!")
            self.region_info.setStyleSheet("background-color: #e3f2fd; padding: 10px; border-radius: 5px;")
        else:
            self.image_viewer.enable_region_selection(False)
            
    def on_region_selected(self, x, y, w, h):
        """Handle region selection"""
        self.region_info.setText(f"✅ Selected region: x={x}, y={y}, width={w}, height={h}\n⏳ Processing...")
        self.region_info.setStyleSheet("background-color: #d4edda; padding: 10px; border-radius: 5px;")
        
        # Auto-process region
        self.process_validation(region=(x, y, w, h))
        
    def process_validation(self, region=None):
        """Process validation"""
        if self.doc_img is None:
            QMessageBox.warning(self, "Warning", "Please upload a document first")
            return
            
        validation_type = self.type_combo.currentText()
        
        # Get keywords for keyword match
        keywords = None
        if validation_type == "Keyword Match":
            text = self.keywords_input.toPlainText().strip()
            if not text:
                QMessageBox.warning(self, "Warning", "Please enter keywords")
                return
            keywords = [k.strip().lower() for k in text.split('\n') if k.strip()]
            
        # Get region for region selection
        if validation_type == "Select Region" and not region:
            QMessageBox.information(self, "Info", "Please select a region on the image")
            return
            
        # Get selfie for face similarity
        if validation_type == "Face Similarity" and self.selfie_img is None:
            QMessageBox.warning(self, "Warning", "Please upload a selfie image")
            return
            
        # Show progress
        self.progress_bar.setVisible(True)
        self.progress_bar.setRange(0, 0)
        self.process_btn.setEnabled(False)
        self.results_text.clear()
        self.results_text.append("Processing... Please wait...\n")
        
        # Create worker thread
        self.worker = OCRWorker(
            validation_type.lower().replace(" ", "_"),
            self.doc_img,
            self.selfie_img if validation_type == "Face Similarity" else None,
            keywords,
            region
        )
        self.worker.progress.connect(self.on_progress_update)
        self.worker.finished.connect(self.on_processing_finished)
        self.worker.start()
        
    def on_progress_update(self, message):
        """Update progress message"""
        self.results_text.append(message)
        self.statusBar().showMessage(message)
        
    def on_processing_finished(self, result):
        """Handle processing results"""
        self.progress_bar.setVisible(False)
        self.process_btn.setEnabled(True)
        self.statusBar().showMessage("Processing complete")
        
        if "error" in result:
            self.results_text.setText(f"❌ Error: {result['error']}")
            QMessageBox.critical(self, "Error", result['error'])
            return
            
        # Display results based on validation type
        validation_type = self.type_combo.currentText()
        
        self.results_text.clear()
        
        if validation_type == "Keyword Match":
            self.results_text.append("=" * 70)
            self.results_text.append("🔍 KEYWORD MATCH RESULTS")
            self.results_text.append("=" * 70)
            self.results_text.append("")
            
            if result["found"]:
                self.results_text.append(f"✅ Found keywords: {', '.join(result['found'])}")
            if result["not_found"]:
                self.results_text.append(f"❌ Missing keywords: {', '.join(result['not_found'])}")
            
            self.results_text.append("")
            self.results_text.append(f"📊 Score: {result['score']}/{result['total']}")
            self.results_text.append(f"📈 Percentage: {result['percentage']:.1f}%")
            self.results_text.append(f"🎯 Status: {'✅ MATCHED' if result['matched'] else '❌ NOT MATCHED'}")
            self.results_text.append("")
            self.results_text.append("=" * 70)
            self.results_text.append("📝 EXTRACTED TEXT (First 50 lines):")
            self.results_text.append("=" * 70)
            self.results_text.append(result.get("all_text", "No text found"))
            
        elif validation_type == "Expiration Date":
            self.results_text.append("=" * 70)
            self.results_text.append("📅 EXPIRATION DATE CHECK")
            self.results_text.append("=" * 70)
            self.results_text.append("")
            status = result["status"]
            if "Valid" in status:
                self.results_text.append(f"✅ {status}")
            elif "Expired" in status:
                self.results_text.append(f"⚠️ {status}")
            else:
                self.results_text.append(f"ℹ️ {status}")
            self.results_text.append("")
            self.results_text.append("=" * 70)
            self.results_text.append("📝 EXTRACTED TEXT (First 50 lines):")
            self.results_text.append("=" * 70)
            self.results_text.append(result.get("all_text", "No text found"))
            
        elif validation_type == "Resolution Check":
            self.results_text.append("=" * 70)
            self.results_text.append("📏 RESOLUTION CHECK")
            self.results_text.append("=" * 70)
            self.results_text.append("")
            self.results_text.append(f"📐 Dimensions: {result['width']} x {result['height']} pixels")
            self.results_text.append(f"🔍 Quality: {result['quality']}")
            if "Low" in result['quality']:
                self.results_text.append("⚠️ Low resolution may affect OCR accuracy")
            
        elif validation_type == "Face Similarity":
            self.results_text.append("=" * 70)
            self.results_text.append("👤 FACE SIMILARITY RESULTS")
            self.results_text.append("=" * 70)
            self.results_text.append("")
            if result["match"]:
                self.results_text.append("✅ VERDICT: Faces MATCH!")
            else:
                self.results_text.append("❌ VERDICT: Faces DO NOT MATCH")
            self.results_text.append("")
            self.results_text.append(f"📊 Confidence: {result['confidence']:.2%}")
            self.results_text.append(f"📏 Distance: {result['distance']:.4f}")
            
        elif validation_type == "Select Region":
            self.results_text.append("=" * 70)
            self.results_text.append("✂️ REGION TEXT EXTRACTION")
            self.results_text.append("=" * 70)
            self.results_text.append("")
            if result["text"]:
                self.results_text.append(f"✅ Extracted {result['words']} words!")
                self.results_text.append("")
                self.results_text.append("📋 Extracted Text:")
                self.results_text.append("-" * 70)
                self.results_text.append(result["text"])
                
                # Copy to clipboard
                clipboard = QApplication.clipboard()
                clipboard.setText(result["text"])
                self.region_info.setText(f"✅ Extracted {result['words']} words! Text copied to clipboard!")
                QMessageBox.information(self, "Success", f"Text extracted and copied to clipboard!\n\n{result['text'][:200]}...")
            else:
                self.results_text.append("❌ No text found in selected region")
                self.region_info.setText("❌ No text found in selected region. Try a different area.")
                
        # Scroll to top
        self.results_text.moveCursor(self.results_text.textCursor().Start)

def main():
    app = QApplication(sys.argv)
    app.setStyle('Fusion')
    window = DocumentValidator()
    window.show()
    sys.exit(app.exec_())

if __name__ == "__main__":
    main()
````

````python
# Keyword validator version-1
import streamlit as st, json
from PIL import Image
from utils import rapid_ocr, check_path, keyword_score, predict_keyword

st.title("📄 OCR Keyword Validator")

file_path = "labels.json"

uploaded_files = st.file_uploader("Upload Images", type=["jpg", "jpeg", "png"], accept_multiple_files=True)

if uploaded_files:
    st.subheader("📸 Image Preview")
    for preview_file in uploaded_files:
        image = Image.open(preview_file)
        st.image(image, caption=preview_file.name, width='content')
        
keyword_input = st.text_input("Enter keywords (comma separated)")
label = st.text_input("Enter label for this document(Ex. Driving License Front)").strip()
threshold = st.slider("Acceptance Threshold (%)", 0, 100, 75)

if "suggestions" not in st.session_state:
    st.session_state.suggestions = None
    
if st.button("Run Validation"):
    keywords = [k.strip() for k in keyword_input.split(",")]
    if not label: st.error("Please enter a label for the document.")
    if not uploaded_files: st.warning("Please upload at least one image.")
    results = []
    texts = []
    for file in uploaded_files:
        ocr_text = rapid_ocr(file)
        texts.append(ocr_text)
        score = keyword_score(ocr_text, keywords)
        print(file.name, score)
        results.append(score)
    avg = sum(results) / len(results) if results else 0
        
    if avg >= threshold:
        st.success("✅ Accepted")
        existing_data = check_path(file_path)
        existing_data[label] = keywords
        with open(file_path, "w") as f: json.dump(existing_data, f, indent=4)
        # st.info(f"Saved '{label}' with keywords: {', '.join(keywords)}")
        st.json({"label": label, "keywords": keywords})
        st.session_state.suggestions = None
    else:
        st.error("❌ Rejected"); st.warning("Document did not meet the acceptance threshold.")
        if st.session_state.suggestions is None:
            st.session_state.suggestions = predict_keyword(texts)

# 👇 always show if exists (prevents disappearing on copy)
    if st.session_state.suggestions:
        st.write("Here are some words you can add:")
        # st.code(st.session_state.suggestions)
        st.write(", ".join(st.session_state.suggestions))
        
    # st.write("### 📊 Results per file")
    # for f, score in zip(uploaded_files, results): st.write(f"{f.name} → {score:.2f}%")

    # st.write("### 📈 Average Score"); st.success(f"{avg:.2f}%")
    
############
# utils.py
############
import re, json, os
from rapidfuzz import process, fuzz
from rapidocr_onnxruntime import RapidOCR
from collections import Counter

ocr = RapidOCR()
file_path = "labels.json"

def normalize(text): 
    # Keep spaces, only remove extra whitespace
    return re.sub(r'\s+', ' ', re.sub(r'[^a-z0-9\s]', '', text.lower())).strip()

def rapid_ocr(file):
    file.seek(0)
    file_bytes = file.read()
    result, _ = ocr(file_bytes)
    return [line[1] for line in result] if result else []

def check_path(file_path):
    if not os.path.exists(file_path): return {}
    try:
        with open(file_path, "r") as f: return json.load(f)
    except: return {}

def keyword_score(ocr_text, fields):
    if not ocr_text or not fields: return 0
    matched = set()
    clean_fields = [normalize(f) for f in fields]
    clean_ocr_text = [normalize(o) for o in ocr_text if o]
    for ocr_word in clean_ocr_text:
        if len(ocr_word) < 3: continue
        match = process.extractOne(ocr_word, clean_fields, scorer=fuzz.partial_ratio)
        if match and match[1] >= 75:
            matched.add(match[0])
    concatenated = ' '.join(clean_ocr_text)
    for field in clean_fields:
        if field in concatenated:
            matched.add(field)
    print(f'matched: {matched}')
    return (len(matched) / len(fields)) * 100 if fields else 0

def cleaned_text(text): return re.sub(r'\s+', ' ', re.sub(r'[^a-z0-9 ]', ' ', text.lower())).strip()

def extract_tokens(text_list):
    tokens = []
    for line in text_list:
        if not line: continue
        clean = cleaned_text(line)
        words = clean.split()
        for i in range(len(words)):
            tokens.append(words[i])
            if i+1 < len(words):
                tokens.append(words[i] + " " + words[i+1])
            if i+2 < len(words):
                tokens.append(words[i] + " " + words[i+1] + " " + words[i+2])
    return tokens

def predict_keyword(texts, threshold=80):
    all_tokens = []
    for text in texts:
        tokens = extract_tokens(text)
        all_tokens.extend(set(tokens))
    
    counter = Counter(all_tokens)
    filtered = [word for word, count in counter.items() if count >= 2 and len(word) > 6]
    
    clusters = []
    for phrase in filtered:
        matched = False
        for cluster in clusters:
            if fuzz.partial_ratio(phrase, cluster[0]) > threshold:
                cluster.append(phrase)
                matched = True
                break
        if not matched: clusters.append([phrase])
    return [max(cluster, key=len) for cluster in clusters]

# Keyword validator version-2

import streamlit as st, json, uuid, datetime
from utils import rapid_ocr, check_path, keyword_score

UUID = str(uuid.uuid4())
output_path = "output.json"

st.title("📄 OCR Check")

uploaded_files = st.file_uploader("Upload Images", type=["jpg", "jpeg", "png"], accept_multiple_files=True)

if st.button("Run Validation"):
    if not uploaded_files: st.warning("Please upload at least one image.");st.stop()
    try:
        with open("labels.json", "r") as f: labels = json.load(f)
    except FileNotFoundError: st.error("labels.json file not found."); st.stop()
    except json.JSONDecodeError: st.error("Invalid JSON in labels.json file.");st.stop()
    
    results = []
    all_file_results = {}
    
    for file in uploaded_files:
        ocr_text = rapid_ocr(file)
        if not ocr_text: results.append({file.name: "No text detected in image"});continue

        doc_scores = {}
        for doc_type, keywords in labels.items():
            score = keyword_score(ocr_text, keywords)
            doc_scores[doc_type] = score
        
        if not doc_scores: results.append({file.name: "No matching document types found"}); continue

        max_score = round(max(doc_scores.values()),2)
        best_matches = [doc_type for doc_type, score in doc_scores.items() if score == max_score]
        
        if len(best_matches) > 1:
            results.append({file.name: {"status": "Multiple document types detected",
                    "matches": best_matches,
                    "scores": {doc: doc_scores[doc] for doc in best_matches}}})
        else: results.append({file.name: {"document_type": best_matches[0],"confidence": max_score}})
        
        all_file_results[file.name] = {
            "document_type": best_matches[0] if len(best_matches) == 1 else "ambiguous",
            "confidence": max_score}
            # "all_scores": doc_scores
        if max_score < 65:
            # results['Suggestions'] = 'Please add more keywords to increase confidence'
            all_file_results[file.name]['Suggestions'] = 'Please add more keywords to increase confidence'
    
    st.subheader("📊 Results")
    for result in results:
        for k, v in result.items():
            try:
                st.write(f"{k} → {v['document_type']} with {v['confidence']}")
            except KeyError:
                st.warning(f"Missing keys in {k}: {v}")

    existing_data = check_path(output_path)
    existing_data[UUID] = {'timestamp': datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S"),'results': all_file_results}
    
    with open(output_path, "w") as f: json.dump(existing_data, f, indent=4)
    
    st.success(f"Results saved with ID: {UUID}")
````

````python
def inspect_key_tree(obj, prefix=""):
    if isinstance(obj, dict):
        keys = list(obj.keys())
        for i, key in enumerate(keys):
            is_last = i == len(keys) - 1
            connector = "└─ " if is_last else "├─ "
            child_pre = "   " if is_last else "│  "
            val = obj[key]
            if isinstance(val, list):
                print(f"{prefix}{connector}{key}  [list]")
                inspect_key_tree(val, prefix + child_pre)
            elif isinstance(val, dict):
                print(f"{prefix}{connector}{key}  {{dict}}")
                inspect_key_tree(val, prefix + child_pre)
            else:
                print(f"{prefix}{connector}{key}")

    elif isinstance(obj, list):
        for i, item in enumerate(obj):
            is_last = i == len(obj) - 1
            connector = "└─ " if is_last else "├─ "
            child_pre = "   " if is_last else "│  "
            if isinstance(item, list):
                print(f"{prefix}{connector}[list]")
                inspect_key_tree(item, prefix + child_pre)
            elif isinstance(item, dict):
                print(f"{prefix}{connector}{{dict}}")
                inspect_key_tree(item, prefix + child_pre)

print("root")
````

````python
# Speech to text
import streamlit as st
from streamlit_mic_recorder import mic_recorder
from faster_whisper import WhisperModel
import tempfile
import os

# ---------------------------
# Load Whisper Model
# ---------------------------
model = WhisperModel(
    "base",
    device="cpu",
    compute_type="int8"
)

# ---------------------------
# Example Keys
# ---------------------------
fields = [
    "Name",
    "Passport Number",
    "Nationality",
    "Date of Birth"
]

# Store results
if "results" not in st.session_state:
    st.session_state.results = {}

st.title("Voice Assisted Key-Value Extraction")

# ---------------------------
# UI Loop
# ---------------------------
for field in fields:

    st.subheader(field)

    audio = mic_recorder(
        start_prompt="🎤 Start Recording",
        stop_prompt="⏹ Stop",
        key=field
    )

    if audio:

        # Save temp audio file
        with tempfile.NamedTemporaryFile(delete=False, suffix=".wav") as tmp_file:
            tmp_file.write(audio["bytes"])
            temp_audio_path = tmp_file.name

        # Transcribe
        segments, info = model.transcribe(temp_audio_path)

        text = " ".join([segment.text for segment in segments])

        # Store result
        st.session_state.results[field] = text

        # Display
        st.success(f"Recognized: {text}")

        # Cleanup
        os.remove(temp_audio_path)

# ---------------------------
# Final Output
# ---------------------------
st.divider()

st.subheader("Final JSON")

st.json(st.session_state.results)
````
